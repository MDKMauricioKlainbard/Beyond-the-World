# Section 3F – Duality
*From Axler, 4th ed.*

### Exercise 3F.1

**Statement:**  
Explain why each linear functional is surjective or is the zero map.

**Solution:**

Let $\varphi : V \to \mathbf{F}$ be a linear functional (a linear map from $V$ to the field of scalars).

- If $\varphi(v) = 0$ for **every** $v \in V$, then by definition $\varphi$ is the zero map.

- Otherwise, there exists some $v_0 \in V$ such that $\varphi(v_0) \neq 0$.  
  For any scalar $\alpha \in \mathbf{F}$, consider the vector  

  $$
  w = \frac{\alpha}{\varphi(v_0)}\, v_0.
  $$

  Then  

  $$
  \varphi(w) = \varphi\!\left(\frac{\alpha}{\varphi(v_0)}\, v_0\right) = \frac{\alpha}{\varphi(v_0)}\, \varphi(v_0) = \alpha.
  $$

  Thus every $\alpha$ is attained as $\varphi(w)$, so $\varphi$ is **surjective**.

Hence a linear functional is either the zero map or surjective. ∎

### Exercise 3F.4

**Statement:**  
Suppose $V$ is finite‑dimensional and $U$ is a subspace of $V$ such that $U \neq V$.  
Prove that there exists $\varphi \in V'$ such that $\varphi(u) = 0$ for every $u \in U$ but $\varphi \neq 0$.

**Solution.**

Since $U \neq V$ and $V$ is finite‑dimensional, we have $\dim U < \dim V$.  
Let $\{u_1,\dots,u_m\}$ be a basis of $U$ (with $m = \dim U$). Extend this list to a basis of $V$:

$$
\{u_1,\dots,u_m, w_1,\dots,w_k\},
$$

where $k = \dim V - \dim U \ge 1$ (because $U \neq V$ implies $k > 0$).

Define a linear functional $\varphi \in V'$ by specifying its values on the basis and extending linearly:

- $\varphi(u_i) = 0$ for $i = 1,\dots,m$,
- $\varphi(w_1) = 1$,
- $\varphi(w_j) = 0$ for $j = 2,\dots,k$.

By the linear map lemma (or by linear extension), this defines a unique linear functional on $V$.

**Properties:**  

- For any $u \in U$, $u$ is a linear combination of the $u_i$'s, so $\varphi(u) = 0$.  
- $\varphi(w_1) = 1 \neq 0$, hence $\varphi$ is not the zero functional.

Thus $\varphi$ satisfies the required conditions. ∎

### Exercise 3F.15

**Statement:**  
Define $T : \mathcal{P}(\mathbb{R}) \to \mathcal{P}(\mathbb{R})$ by  

$$
(Tp)(x) = x^2 p(x) + p''(x) \qquad (x \in \mathbb{R}).
$$

(a) Suppose $\varphi \in \mathcal{P}(\mathbb{R})'$ is defined by $\varphi(p) = p'(4)$. Describe the linear functional $T'(\varphi)$ on $\mathcal{P}(\mathbb{R})$.

(b) Suppose $\varphi \in \mathcal{P}(\mathbb{R})'$ is defined by $\varphi(p) = \int_0^1 p$. Evaluate $(T'(\varphi))(x^3)$.

---

**Solution.**

Recall that for a linear map $T$, the dual map $T'$ satisfies  

$$
(T'(\varphi))(p) = \varphi(Tp) \qquad (\varphi \in \mathcal{P}(\mathbb{R})',\; p \in \mathcal{P}(\mathbb{R})).
$$

---

#### Part (a)

We have $\varphi(p) = p'(4)$. Then  

$$
(T'(\varphi))(p) = \varphi(Tp) = (Tp)'(4).
$$

First compute the derivative of $Tp$:

$$
(Tp)(x) = x^2 p(x) + p''(x),
$$
$$
(Tp)'(x) = 2x p(x) + x^2 p'(x) + p'''(x).
$$

Evaluating at $x = 4$:

$$
(Tp)'(4) = 8 p(4) + 16 p'(4) + p'''(4).
$$

Therefore  

$$
T'(\varphi)(p) = 8p(4) + 16p'(4) + p'''(4).
$$

---

#### Part (b)

Now $\varphi(p) = \int_0^1 p(x) dx$. Then  

$$
(T'(\varphi))(x^3) = \varphi(T(x^3)).
$$

Compute $T(x^3)$:

$$
T(x^3) = x^2 \cdot x^3 + (x^3)'' = x^5 + (6x) = x^5 + 6x.
$$

Now apply $\varphi$:

$$
\varphi(T(x^3)) = \int_0^1 (x^5 + 6x) dx = \int_0^1 x^5 dx + 6 \int_0^1 x dx.
$$

Compute each integral:

$$
\int_0^1 x^5 dx = \left[ \frac{x^6}{6} \right]_0^1 = \frac{1}{6},
$$
$$
\int_0^1 x dx = \left[ \frac{x^2}{2} \right]_0^1 = \frac{1}{2}.
$$

Thus  

$$
\varphi(T(x^3)) = \frac{1}{6} + 6 \cdot \frac{1}{2} = \frac{1}{6} + 3 = \frac{19}{6}.
$$

Hence  

$$
(T'(\varphi))(x^3) = \frac{19}{6}.
$$

∎
