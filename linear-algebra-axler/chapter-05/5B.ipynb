## Exercise 5.B.1

**Statement.**

Suppose $T \in \mathcal{L}(V)$. Prove that $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$.


### Solution

#### ($\Rightarrow$)

Assume $9$ is an eigenvalue of $T^2$. Then there exists a non‑zero vector $v \in V$ such that $T^2 v = 9 v$. This can be rewritten as  

$$
(T^2 - 9I)v = 0 \quad\Longrightarrow\quad (T - 3I)(T + 3I)v = 0.
$$

Let $w = (T + 3I)v$.  

- If $w = 0$, then $(T + 3I)v = 0$, so $-3$ is an eigenvalue of $T$ (with eigenvector $v$).  
- If $w \neq 0$, then $(T - 3I)w = 0$, which means $3$ is an eigenvalue of $T$ (with eigenvector $w$).

In either case, $3$ or $-3$ is an eigenvalue of $T$.

---

#### ($\Leftarrow$)

Suppose $3$ or $-3$ is an eigenvalue of $T$.

- If $3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = 3v$. Then $T^2 v = T(3v) = 3 Tv = 9v$, so $9$ is an eigenvalue of $T^2$.
- If $-3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = -3v$. Then $T^2 v = T(-3v) = -3 Tv = (-3)(-3v) = 9v$, so again $9$ is an eigenvalue of $T^2$.

Thus $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$. ∎

## Exercise 5.B.2

**Statement.**

Suppose $V$ is a complex vector space and $T \in \mathcal{L}(V)$ has no eigenvalues. Prove that every subspace of $V$ invariant under $T$ is either $\{0\}$ or infinite‑dimensional.


### Solution

Let $U \subseteq V$ be a subspace invariant under $T$. Assume that $U$ is finite‑dimensional and $U \neq \{0\}$.  

Because $V$ is a complex vector space, $U$ (as a subspace) is also a complex vector space. The restriction $T|_U : U \to U$ is a linear operator on a non‑zero finite‑dimensional complex vector space. By the fundamental theorem of algebra (or the standard theorem that every operator on a finite‑dimensional complex vector space has an eigenvalue), $T|_U$ possesses an eigenvalue $\lambda \in \mathbb{C}$. Hence there exists a non‑zero $u \in U$ such that $T|_U(u) = \lambda u$, i.e. $Tu = \lambda u$.  

Then $\lambda$ is an eigenvalue of $T$ (with eigenvector $u$), contradicting the hypothesis that $T$ has no eigenvalues.

Therefore such a $U$ cannot exist. Thus any invariant subspace $U$ must be either $\{0\}$ or infinite‑dimensional. ∎

## Exercise 5.B.3

**Statement.**

Suppose $n$ is a positive integer and $T \in \mathcal{L}(\mathbb{F}^n)$ is defined by  

$$
T(x_1, \dots, x_n) = (x_1 + \dots + x_n, \dots, x_1 + \dots + x_n).
$$

(a) Find all eigenvalues and eigenvectors of $T$.  
(b) Find the minimal polynomial of $T$.


### Solution

Choose the standard basis $\{e_1,\dots,e_n\}$ of $\mathbb{F}^n$. For each $e_k$ we have  

$$
T(e_k) = (1,1,\dots,1) = e_1 + e_2 + \dots + e_n.
$$

Thus the matrix of $T$ with respect to this basis is the $n \times n$ matrix $A$ whose every entry is $1$.

---

#### (a) Eigenvalues and eigenvectors

First compute $A^2$. The $(i,j)$-entry of $A^2$ is the dot product of the $i$-th row (all ones) with the $j$-th column (all ones), which equals $n$. Hence $A^2 = nA$.

Now suppose $\lambda$ is an eigenvalue of $A$ with eigenvector $v \neq 0$. Then $Av = \lambda v$. Applying $A$ again gives $A^2 v = \lambda^2 v$. But $A^2 v = nA v = n \lambda v$. Therefore $\lambda^2 v = n \lambda v$, i.e. $(\lambda^2 - n\lambda)v = 0$. Since $v \neq 0$, we obtain $\lambda(\lambda - n) = 0$, so $\lambda = 0$ or $\lambda = n$.

- **Eigenvalue $n$:**  
  The vector $v = (1,1,\dots,1)$ satisfies $Av = (n,n,\dots,n) = n v$. Hence every non‑zero scalar multiple of $(1,\dots,1)$ is an eigenvector.

- **Eigenvalue $0$:**  
  $Av = 0$ means that all components of $Av$ are zero, i.e. $x_1 + \dots + x_n = 0$. The set $\{ (x_1,\dots,x_n) : x_1+\dots+x_n = 0 \}$ is a subspace of dimension $n-1$. Any non‑zero vector in this subspace is an eigenvector with eigenvalue $0$.

Thus the eigenvalues are $0$ and $n$, and the eigenvectors are exactly the non‑zero vectors with sum $0$ (for $0$) and the non‑zero multiples of $(1,\dots,1)$ (for $n$).

---

#### (b) Minimal polynomial

We have $A^2 - nA = 0$, so the monic polynomial $p(z) = z^2 - nz = z(z-n)$ satisfies $p(A)=0$.

- If a polynomial of degree $1$, say $z - c$, satisfied $A - cI = 0$, then $A$ would be $cI$, which it is not.
- Therefore the minimal polynomial cannot be of degree $1$. Since $p$ is degree $2$ and annihilates $A$, it must be the minimal polynomial.

Hence the minimal polynomial of $T$ is $z(z-n)$. ∎
