# Section 5B – The Minimal Polynomial

*From Axler, 4th ed.*

### Exercise 5B.1

**Statement:**  
Suppose $T \in \mathcal{L}(V)$. Prove that $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$.

**Proof.**

---

#### ($\Rightarrow$)  
Assume $9$ is an eigenvalue of $T^2$. Then there exists a non‑zero vector $v \in V$ such that $T^2 v = 9 v$. This can be rewritten as  

$$
(T^2 - 9I)v = 0 \quad\Longrightarrow\quad (T - 3I)(T + 3I)v = 0.
$$

Let $w = (T + 3I)v$.  

- If $w = 0$, then $(T + 3I)v = 0$, so $-3$ is an eigenvalue of $T$ (with eigenvector $v$).  
- If $w \neq 0$, then $(T - 3I)w = 0$, which means $3$ is an eigenvalue of $T$ (with eigenvector $w$).

In either case, $3$ or $-3$ is an eigenvalue of $T$.

---

#### ($\Leftarrow$)  
Suppose $3$ or $-3$ is an eigenvalue of $T$.

- If $3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = 3v$. Then $T^2 v = T(3v) = 3 Tv = 9v$, so $9$ is an eigenvalue of $T^2$.
- If $-3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = -3v$. Then $T^2 v = T(-3v) = -3 Tv = (-3)(-3v) = 9v$, so again $9$ is an eigenvalue of $T^2$.

Thus $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$. ∎
