## Exercise 5.B.1

**Statement.**

Suppose $T \in \mathcal{L}(V)$. Prove that $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$.


### Solution

#### ($\Rightarrow$)

Assume $9$ is an eigenvalue of $T^2$. Then there exists a non‑zero vector $v \in V$ such that $T^2 v = 9 v$. This can be rewritten as  

$$
(T^2 - 9I)v = 0 \quad\Longrightarrow\quad (T - 3I)(T + 3I)v = 0.
$$

Let $w = (T + 3I)v$.  

- If $w = 0$, then $(T + 3I)v = 0$, so $-3$ is an eigenvalue of $T$ (with eigenvector $v$).  
- If $w \neq 0$, then $(T - 3I)w = 0$, which means $3$ is an eigenvalue of $T$ (with eigenvector $w$).

In either case, $3$ or $-3$ is an eigenvalue of $T$.

---

#### ($\Leftarrow$)

Suppose $3$ or $-3$ is an eigenvalue of $T$.

- If $3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = 3v$. Then $T^2 v = T(3v) = 3 Tv = 9v$, so $9$ is an eigenvalue of $T^2$.
- If $-3$ is an eigenvalue, there exists a non‑zero $v$ with $Tv = -3v$. Then $T^2 v = T(-3v) = -3 Tv = (-3)(-3v) = 9v$, so again $9$ is an eigenvalue of $T^2$.

Thus $9$ is an eigenvalue of $T^2$ if and only if $3$ or $-3$ is an eigenvalue of $T$. ∎

## Exercise 5.B.2

**Statement.**

Suppose $V$ is a complex vector space and $T \in \mathcal{L}(V)$ has no eigenvalues. Prove that every subspace of $V$ invariant under $T$ is either $\{0\}$ or infinite‑dimensional.


### Solution

Let $U \subseteq V$ be a subspace invariant under $T$. Assume that $U$ is finite‑dimensional and $U \neq \{0\}$.  

Because $V$ is a complex vector space, $U$ (as a subspace) is also a complex vector space. The restriction $T|_U : U \to U$ is a linear operator on a non‑zero finite‑dimensional complex vector space. By the fundamental theorem of algebra (or the standard theorem that every operator on a finite‑dimensional complex vector space has an eigenvalue), $T|_U$ possesses an eigenvalue $\lambda \in \mathbb{C}$. Hence there exists a non‑zero $u \in U$ such that $T|_U(u) = \lambda u$, i.e. $Tu = \lambda u$.  

Then $\lambda$ is an eigenvalue of $T$ (with eigenvector $u$), contradicting the hypothesis that $T$ has no eigenvalues.

Therefore such a $U$ cannot exist. Thus any invariant subspace $U$ must be either $\{0\}$ or infinite‑dimensional. ∎

## Exercise 5.B.3

**Statement.**

Suppose $n$ is a positive integer and $T \in \mathcal{L}(\mathbb{F}^n)$ is defined by  

$$
T(x_1, \dots, x_n) = (x_1 + \dots + x_n, \dots, x_1 + \dots + x_n).
$$

(a) Find all eigenvalues and eigenvectors of $T$.  
(b) Find the minimal polynomial of $T$.


### Solution

Choose the standard basis $\{e_1,\dots,e_n\}$ of $\mathbb{F}^n$. For each $e_k$ we have  

$$
T(e_k) = (1,1,\dots,1) = e_1 + e_2 + \dots + e_n.
$$

Thus the matrix of $T$ with respect to this basis is the $n \times n$ matrix $A$ whose every entry is $1$.

---

#### (a) Eigenvalues and eigenvectors

First compute $A^2$. The $(i,j)$-entry of $A^2$ is the dot product of the $i$-th row (all ones) with the $j$-th column (all ones), which equals $n$. Hence $A^2 = nA$.

Now suppose $\lambda$ is an eigenvalue of $A$ with eigenvector $v \neq 0$. Then $Av = \lambda v$. Applying $A$ again gives $A^2 v = \lambda^2 v$. But $A^2 v = nA v = n \lambda v$. Therefore $\lambda^2 v = n \lambda v$, i.e. $(\lambda^2 - n\lambda)v = 0$. Since $v \neq 0$, we obtain $\lambda(\lambda - n) = 0$, so $\lambda = 0$ or $\lambda = n$.

- **Eigenvalue $n$:**  
  The vector $v = (1,1,\dots,1)$ satisfies $Av = (n,n,\dots,n) = n v$. Hence every non‑zero scalar multiple of $(1,\dots,1)$ is an eigenvector.

- **Eigenvalue $0$:**  
  $Av = 0$ means that all components of $Av$ are zero, i.e. $x_1 + \dots + x_n = 0$. The set $\{ (x_1,\dots,x_n) : x_1+\dots+x_n = 0 \}$ is a subspace of dimension $n-1$. Any non‑zero vector in this subspace is an eigenvector with eigenvalue $0$.

Thus the eigenvalues are $0$ and $n$, and the eigenvectors are exactly the non‑zero vectors with sum $0$ (for $0$) and the non‑zero multiples of $(1,\dots,1)$ (for $n$).

---

#### (b) Minimal polynomial

We have $A^2 - nA = 0$, so the monic polynomial $p(z) = z^2 - nz = z(z-n)$ satisfies $p(A)=0$.

- If a polynomial of degree $1$, say $z - c$, satisfied $A - cI = 0$, then $A$ would be $cI$, which it is not.
- Therefore the minimal polynomial cannot be of degree $1$. Since $p$ is degree $2$ and annihilates $A$, it must be the minimal polynomial.

Hence the minimal polynomial of $T$ is $z(z-n)$. ∎

## Exercise 5.B.6

**Statement.**

Suppose $T \in \mathcal{L}(\mathbb{F}^2)$ is defined by $T(w, z) = (-z, w)$. Find the minimal polynomial of $T$.


### Solution

Compute $T^2$:

$$
T^2(w,z) = T(T(w,z)) = T(-z, w) = (-w, -z) = -(w, z).
$$

Thus $T^2(w,z) = - (w,z)$ for all $(w,z) \in \mathbb{F}^2$, i.e. $T^2 + I = 0$.

Therefore the polynomial $p(z) = z^2 + 1$ satisfies $p(T) = 0$.

- If a polynomial of degree $1$, say $z - c$, annihilated $T$, then $T$ would be $cI$. But $T$ is not a scalar multiple of the identity (e.g. $T(1,0) = (0,1)$ is not a multiple of $(1,0)$).
- Hence no linear polynomial works.

Thus the minimal polynomial is $z^2 + 1$. ∎


### Exercise 5B.8

**Statement:**  
Suppose $T \in \mathcal{L}(\mathbb{R}^2)$ is the operator of counterclockwise rotation by $1^\circ$. Find the minimal polynomial of $T$.

**Solution.**

Let $\theta = 1^\circ$ (in radians, but we keep degrees).  
Take the standard basis $\{e_1, e_2\}$ of $\mathbb{R}^2$. The matrix of $T$ in this basis is  

$$
A = \begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}.
$$

Since $\dim \mathbb{R}^2 = 2$, the minimal polynomial of $T$ has degree at most $2$.

- It cannot be of degree $1$ because $T$ is not a scalar multiple of the identity (a rotation by $\theta$ with $\theta \not\equiv 0 \pmod{180^\circ}$ does not map every vector to a multiple of itself).  
- Hence the minimal polynomial must be quadratic: $p(z) = z^2 + \alpha z + \beta$ with $\alpha, \beta \in \mathbb{R}$ (monic).

We determine $\alpha$ and $\beta$ by applying $p(T)$ to a convenient non‑zero vector, for instance $e_1 = (1,0)$.  
Compute:

$$
T e_1 = (\cos\theta, \sin\theta), \qquad
T^2 e_1 = T(\cos\theta, \sin\theta) = (\cos 2\theta, \sin 2\theta).
$$

The condition $p(T)=0$ means $T^2 e_1 + \alpha T e_1 + \beta e_1 = 0$. In coordinates:

$$
\begin{cases}
\cos 2\theta + \alpha \cos\theta + \beta = 0,\\[2pt]
\sin 2\theta + \alpha \sin\theta = 0.
\end{cases}
$$

From the second equation, $\alpha = -\dfrac{\sin 2\theta}{\sin\theta}$. Using $\sin 2\theta = 2\sin\theta\cos\theta$, we obtain $\alpha = -2\cos\theta$.

Substitute into the first equation:

$$
\cos 2\theta + (-2\cos\theta)\cos\theta + \beta = 0
\;\Longrightarrow\; \cos 2\theta - 2\cos^2\theta + \beta = 0.
$$

Recall $\cos 2\theta = 2\cos^2\theta - 1$. Thus

$$
(2\cos^2\theta - 1) - 2\cos^2\theta + \beta = 0 \;\Longrightarrow\; -1 + \beta = 0 \;\Longrightarrow\; \beta = 1.
$$

Therefore the monic polynomial $p(z) = z^2 - 2\cos\theta \, z + 1$ satisfies $p(T)=0$. Because no linear polynomial annihilates $T$, this quadratic is the minimal polynomial.

Hence, for $\theta = 1^\circ$,

$$
\boxed{z^2 - 2\cos 1^\circ\; z + 1}.
$$

∎

### Exercise 5B.12

**Statement:**  
Define $T \in \mathcal{L}(\mathbb{F}^n)$ by $T(x_1, x_2, x_3, \dots, x_n) = (x_1, 2x_2, 3x_3, \dots, n x_n)$. Find the minimal polynomial of $T$.

**Solution.**

Take the standard basis $\{e_1, e_2, \dots, e_n\}$ of $\mathbb{F}^n$, where $e_k$ has $1$ in the $k$-th coordinate and $0$ elsewhere.  
For each $k$, we have  

$$
T e_k = k e_k.
$$

Thus $1,2,\dots,n$ are eigenvalues of $T$, each with eigenvector $e_k$.

A polynomial $p(z)$ annihilates $T$ iff $p(k)=0$ for every $k=1,\dots,n$ (since $p(T)e_k = p(k)e_k$). Therefore $p$ must have $z-1, z-2, \dots, z-n$ as factors. Consequently, the **minimal polynomial** of $T$ (the monic polynomial of smallest degree that annihilates $T$) must be divisible by  

$$
q(z) = (z-1)(z-2)\cdots(z-n).
$$

The degree of $q$ is $n$. Because $T$ acts on an $n$-dimensional space, its minimal polynomial has degree at most $n$. Hence the minimal polynomial is exactly $q(z)$ (it is monic and degree $n$, and no lower degree polynomial can vanish at all $n$ distinct points).  

Thus the minimal polynomial of $T$ is  

$$
\boxed{(z-1)(z-2)\cdots(z-n)}.
$$

∎

## Exercise 5.B.14

**Statement.**

Suppose $V$ is finite‑dimensional and $T \in \mathcal{L}(V)$ has minimal polynomial  

$$
4 + 5z - 6z^2 - 7z^3 + 2z^4 + z^5.
$$

Find the minimal polynomial of $T^{-1}$.


### Solution

Because the constant term $4 \neq 0$, the operator $T$ is invertible.  
Let  

$$
p(z) = a_0 + a_1 z + a_2 z^2 + a_3 z^3 + a_4 z^4 + a_5 z^5,
$$

where $a_0 = 4$, $a_1 = 5$, $a_2 = -6$, $a_3 = -7$, $a_4 = 2$, $a_5 = 1$.  
We have $p(T) = 0$.

Multiply the equality $p(T) = 0$ by $T^{-5}$ (recall $T$ is invertible):

$$
a_0 T^{-5} + a_1 T^{-4} + a_2 T^{-3} + a_3 T^{-2} + a_4 T^{-1} + a_5 I = 0.
$$

This shows that the polynomial  

$$
r(z) = a_0 z^5 + a_1 z^4 + a_2 z^3 + a_3 z^2 + a_4 z + a_5
$$

satisfies $r(T^{-1}) = 0$. In other words, the reversed polynomial of $p$ (scaled) annihilates $T^{-1}$.

To obtain a **monic** polynomial, we divide by the leading coefficient $a_0 = 4$:

$$
q(z) = \frac{1}{a_0} r(z) = z^5 + \frac{a_1}{a_0} z^4 + \frac{a_2}{a_0} z^3 + \frac{a_3}{a_0} z^2 + \frac{a_4}{a_0} z + \frac{a_5}{a_0}.
$$

Substituting the coefficients:

$$
q(z) = z^5 + \frac{5}{4} z^4 + \frac{-6}{4} z^3 + \frac{-7}{4} z^2 + \frac{2}{4} z + \frac{1}{4}
= z^5 + \frac{5}{4}z^4 - \frac{3}{2}z^3 - \frac{7}{4}z^2 + \frac{1}{2}z + \frac{1}{4}.
$$

The polynomial $q(z)$ is monic and annihilates $T^{-1}$. Any polynomial of lower degree that annihilates $T^{-1}$ would, after multiplying by a suitable power of $T$, give a polynomial of lower degree that annihilates $T$, contradicting the minimality of $p(z)$. Hence $q(z)$ is the minimal polynomial of $T^{-1}$.

Therefore the minimal polynomial of $T^{-1}$ is  

$$
z^5 + \frac{5}{4}z^4 - \frac{3}{2}z^3 - \frac{7}{4}z^2 + \frac{1}{2}z + \frac{1}{4}.
$$

∎
