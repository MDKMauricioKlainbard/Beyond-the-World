## Exercise 5.A.38

**Statement.**

Suppose $V$ is finite-dimensional, $T \in \mathcal{L}(V)$, and $U$ is a subspace of $V$ invariant under $T$. The *quotient operator* $T/U \in \mathcal{L}(V/U)$ is defined by  

$$
(T/U)(v + U) = Tv + U
$$

for each $v \in V$.

(a) Show that the definition of $T/U$ makes sense (which requires using the condition that $U$ is invariant under $T$) and show that $T/U$ is an operator on $V/U$.

(b) Show that each eigenvalue of $T/U$ is an eigenvalue of $T$.


### Solution

#### Part (a)

**Well‑definedness.**  
Take $u,v \in V$ such that $v+U = u+U$. Then $v-u \in U$. Because $U$ is invariant under $T$, $T(v-u) \in U$. Now compute  

$$
(T/U)(v+U) - (T/U)(u+U) = (Tv+U) - (Tu+U) = (Tv - Tu) + U = T(v-u) + U.
$$

Since $T(v-u) \in U$, we have $T(v-u)+U = U$, the zero element of $V/U$. Therefore  

$$
(T/U)(v+U) = (T/U)(u+U),
$$

so the definition does not depend on the representative of the coset.

**Linearity.**  
For any $v,w \in V$ and $\lambda \in \mathbf{F}$:

$$
\begin{aligned}
(T/U)\bigl((v+U)+(w+U)\bigr) &= (T/U)\bigl((v+w)+U\bigr) = T(v+w)+U = (Tv+Tw)+U \\
&= (Tv+U)+(Tw+U) = (T/U)(v+U) + (T/U)(w+U),
\end{aligned}
$$

$$
\begin{aligned}
(T/U)\bigl(\lambda(v+U)\bigr) &= (T/U)(\lambda v+U) = T(\lambda v)+U = \lambda Tv+U \\
&= \lambda (Tv+U) = \lambda (T/U)(v+U).
\end{aligned}
$$

Hence $T/U$ is a linear operator on $V/U$.

---

#### Part (b)

Let $\lambda$ be an eigenvalue of $T/U$. Then there exists a non‑zero coset $v+U \in V/U$ such that  

$$
(T/U)(v+U) = \lambda (v+U).
$$

By definition, this means  

$$
Tv + U = \lambda v + U \quad\Longrightarrow\quad Tv - \lambda v \in U.
$$

Thus $(T-\lambda I)v \in U$.

Consider the restriction of $T-\lambda I$ to the subspace $U$. We distinguish two cases.

**Case 1:** $T-\lambda I|_U$ is **not** injective.  
Then there exists a non‑zero $u \in U$ with $(T-\lambda I)u = 0$. This $u$ is an eigenvector of $T$ with eigenvalue $\lambda$, so $\lambda$ is an eigenvalue of $T$.

**Case 2:** $T-\lambda I|_U$ is **injective**.  
Because $U$ is finite‑dimensional (since $V$ is finite‑dimensional), injectivity implies surjectivity. Hence $T-\lambda I|_U$ is also surjective onto $U$.

Since $(T-\lambda I)v \in U$, surjectivity gives a vector $w \in U$ such that  

$$
(T-\lambda I)w = (T-\lambda I)v.
$$

Define $v' = v - w$. If $v' = 0$, then $v = w \in U$ (because $w \in U$), which would imply $v+U = U$, contradicting that $v+U$ is non‑zero. Hence $v' \neq 0$. Moreover,

$$
(T-\lambda I)v' = (T-\lambda I)(v-w) = (T-\lambda I)v - (T-\lambda I)w = 0.
$$

Thus $v'$ is a non‑zero vector in $\ker(T-\lambda I)$. Therefore $\lambda$ is an eigenvalue of $T$.

In both cases, $\lambda$ is an eigenvalue of $T$. ∎
