## Exercise 3.1 — Plotting experimental data

In the on-line resources, you will find a file called `sunspots.txt`, which contains the observed number of sunspots on the Sun for each month since January 1749. The file contains two columns of numbers, the first being the month and the second being the sunspot number.

---

- **(a)** Write a program that reads in the data and makes a graph of sunspots as a function of time.
- **(b)** Modify your program to display only the first 1000 data points on the graph.
- **(c)** Modify your program further to calculate and plot the running average of the data, defined by:

$$Y_k = \frac{1}{2r + 1} \sum_{m=-r}^{r} y_{k+m}$$

where $r = 5$ in this case (and the $y_k$ are the sunspot numbers). Have the program plot both the original data and the running average on the same graph, again over the range covered by the first 1000 data points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Item a
data = np.loadtxt("sunspots.txt")
x = data[:,0]
y = data[:,1]
plt.plot(x,y)
plt.xlabel("Month")
plt.ylabel("Number of sunspots")
plt.show()

# Item b
plt.plot(x[:1000], y[:1000])
plt.xlabel("Month")
plt.ylabel("Number of sunspots")
plt.show()

# Item c
Yk = np.empty(y.size)
r = 5
for m in range (r,y.size-r):
    Yk[m] = 1/(2*r+1) * np.sum(y[m-r:m+r+1])

plt.plot(x[:1000], y[:1000], "k-")
plt.plot(x[:1000], Yk[:1000], "g-")
plt.xlabel("Month")
plt.ylabel("Number of sunspots")
plt.show()
```

## Exercise 3.2 — Curve plotting

Although the `plot` function is designed primarily for plotting standard $xy$ graphs, it can be adapted for other kinds of plotting as well.

---

- **(a)** Make a plot of the so-called *deltoid curve*, which is defined parametrically by the equations

$$
x = 2\cos\theta + \cos 2\theta
$$

$$
y = 2\sin\theta - \sin 2\theta
$$

where $0 \leq \theta < 2\pi$.

Take a set of values of $\theta$ between zero and $2\pi$ and calculate $x$ and $y$ for each from the equations above, then plot $y$ as a function of $x$.

---

- **(b)** Taking this approach a step further, one can make a polar plot $r = f(\theta)$ for some function $f$ by calculating $r$ for a range of values of $\theta$ and then converting $r$ and $\theta$ to Cartesian coordinates using the standard equations

$$
x = r\cos\theta, \qquad y = r\sin\theta
$$

Use this method to make a plot of the Galilean spiral

$$
r = \theta^2
$$

for $0 \leq \theta \leq 10\pi$.

---

- **(c)** Using the same method, make a polar plot of “Fey’s function”

$$
r = e^{\cos\theta} - 2\cos 4\theta + \sin^5\left(\frac{\theta}{12}\right)
$$

in the range

$$
0 \leq \theta \leq 24\pi
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Item a
angle = np.linspace(0, 2*np.pi, 100)
x = 2*np.cos(angle)+np.cos(2*angle)
y = 2*np.sin(angle)-np.sin(2*angle)
plt.plot(x,y, "g-")
plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.title("Deltoid curve")
plt.show()

# Item b
angle = np.linspace(0, 10*np.pi, 500)
r = angle**2
x = r * np.cos(angle)
y = r * np.sin(angle)
plt.plot(x,y, "r-")
plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.title("Galilean spiral")
plt.show()

# Item c
angle = np.linspace(0,24*np.pi, 4000)
r = np.exp(np.cos(angle))-2*np.cos(4*angle)+np.sin(angle/12)**5
x = r*np.cos(angle)
y = r*np.sin(angle)
plt.plot(x,y, "b-")
plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.title("Fey's function")
plt.show()
```

## Exercise 3.3 — Density plots

There is a file in the on-line resources called `stm.txt`, which contains a grid of values from scanning tunneling microscope measurements of the $(111)$ surface of silicon.

A scanning tunneling microscope (STM) is a device that measures the shape of a surface at the atomic level by tracking a sharp tip over the surface and measuring quantum tunneling current as a function of position. The end result is a grid of values that represent the height of the surface and the file `stm.txt` contains just such a grid of values.

Write a program that reads the data contained in the file and makes a density plot of the values. Use the various options and variants you have learned about to make a picture that shows the structure of the silicon surface clearly.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("stm.txt", float)
plt.imshow(data, origin="lower", interpolation="nearest", cmap="viridis")
plt.colorbar(label="Height")
plt.title("STM image of silicon surface")
plt.show()
```
