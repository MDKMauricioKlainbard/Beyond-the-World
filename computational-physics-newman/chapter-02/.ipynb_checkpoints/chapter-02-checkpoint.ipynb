## Exercise 2.2

A satellite is to be launched into a circular orbit around the Earth so that it orbits the planet once every *T* seconds. The altitude *h* above the Earth's surface is:

$$h = \left(\frac{GMT^2}{4\pi^2}\right)^{1/3} - R$$

where $G = 6.67 \times 10^{-11}$ m³ kg⁻¹ s⁻², $M = 5.97 \times 10^{24}$ kg and $R = 6371$ km.

- **(a)** Ask the user for *T* and print the altitude in meters.
- **(b)** Use the program for geosynchronous orbit (1 day), 90 min, and 45 min.
- **(d)** A geosynchronous satellite uses a sidereal day (23.93 h), not 24 h — how much does the altitude change?

**Notes:** The 45-minute case gives a negative altitude (impossible). Sidereal vs solar day differs by ~80 km.


```python
from math import pi


G = 6.67e-11
M = 5.97e24
R = 6371*1000

T = float(input("Enter the desired value of T, in seconds, for the satellite: "))
h = ((G*M*(T)**2)/(4*pi**2))**(1/3)-R
print(f"The height of the satellite must be: {h}m")
print(f"Or roughly: {h / 1000:.2f} kilometers")
```

## Exercise 2.3

Inverse of Example 2.2: ask for Cartesian coordinates *x*, *y* and print polar coordinates with θ in degrees.


```python
from math import atan, pi, sqrt


x = float(input("Enter x-coordinate: "))
y = float(input("Enter y-coordinate: "))
r = sqrt(x**2+y**2)
theta = 0.0

if x == 0:
    if y > 0:
        theta = 90
    elif y < 0:
        theta = 270
    else:
        theta = 0
elif x < 0:
    theta = 180 + atan(y/x) * 180/pi
else:
    theta = atan(y/x)*180/pi

print(f"Polar coordinates: ({r:.2f}, {theta:.2f})")
```

## Exercise 2.5 — Quantum potential step

Particle mass $m = 9.11 \times 10^{-31}$ kg, energy $E = 10$ eV, step height $V = 9$ eV.

$$k_1 = \frac{\sqrt{2mE}}{\hbar}, \quad k_2 = \frac{\sqrt{2m(E-V)}}{\hbar}$$

$$T = \frac{4 k_1 k_2}{(k_1 + k_2)^2}, \quad R = \left(\frac{k_1 - k_2}{k_1 + k_2}\right)^2$$


```python
from math import sqrt


m = 9.11e-31 # Electron mass
E = 10 # Initial electron's kinetic energy measured in eV.
V = 9 # Potential step measured in eV.
h = 6.63e-34 # Planck's constant in m^2kg/s
k1 = sqrt(2*m*E)/h
k2 = sqrt(2*m*(E-V))/h
T = (4*k1*k2)/(k1+k2)**2
R = ((k1-k2)/(k1+k2))**2

print(f"Probability of transmission: {T:.4f}")
print(f"Probability of reflection: {R:.4f}")
print(f"Total probability: {T+R:.4f}") # Should always equal 1.0
```

## Exercise 2.6 — Planetary orbits

Given perihelion distance $l_1$ and velocity $v_1$, compute aphelion $l_2$, $v_2$, period $T$, and eccentricity $e$.

Conservation gives $v_2$ as the smaller root of:

$$v_2^2 - \frac{2GM}{v_1 l_1} v_2 - \left[v_1^2 - \frac{2GM}{l_1}\right] = 0$$

Then $l_2 = l_1 v_1 / v_2$, $a = (l_1+l_2)/2$, $b = \sqrt{l_1 l_2}$, $T = 2\pi a b / (l_1 v_1)$, $e = (l_2-l_1)/(l_2+l_1)$.

Test with Earth ($l_1 = 1.4710 \times 10^{11}$ m, $v_1 = 3.0287 \times 10^4$ m/s) and Halley's comet.


```python
from math import pi, sqrt


M = 1.9891e30 # Mass of the sun (kg)
G = 6.6738e-11 # Newton's gravitational constant (m^3kg^-1s^-2)

v1_earth = 3.0287e4
l1_earth = 1.4710e11

v1_halley = 5.4529e4
l1_halley = 8.7830e10
# Solve the quadratic formula for the smaller root:
# a = 1 for both bodies
b_earth = -2*G*M/(v1_earth*l1_earth)
c_earth = -(v1_earth**2 - 2*G*M/l1_earth)

b_halley = -2*G*M/(v1_halley*l1_halley)
c_halley = -(v1_halley**2 - 2*G*M/l1_halley)

v2_earth = (-b_earth-sqrt(b_earth**2-4*c_earth))/2
v2_halley = (-b_halley-sqrt(b_halley**2-4*c_halley))/2

l2_earth = (l1_earth*v1_earth)/v2_earth
l2_halley = (l1_halley*v1_halley)/v2_halley

major_axis_earth = (l1_earth+l2_earth)/2
major_axis_halley = (l1_halley+l2_halley)/2

minor_axis_earth = sqrt(l1_earth*l2_earth)
minor_axis_halley = sqrt(l1_halley*l2_halley)

T_earth = (2*pi*major_axis_earth*minor_axis_earth)/(l1_earth*v1_earth)
T_halley = (2*pi*major_axis_halley*minor_axis_halley)/(l1_halley*v1_halley)

e_earth = (l2_earth-l1_earth)/(l2_earth+l1_earth)
e_halley = (l2_halley-l1_halley)/(l2_halley+l1_halley)

conversion_seconds_to_years = 1/31536000
conversion_meters_to_kilometers = 1/1000

print("Earth data")
print(f"Orbital period in years: {T_earth*conversion_seconds_to_years}")
print(f"Velocity at aphelion: {v2_earth:.2f}m/s")
print(f"Distance at the sun in aphelion: {l2_earth*conversion_meters_to_kilometers:.2f}km")
print(f"Excentricity: {e_earth:.4f}")
print("---")
print("Halley's comet data")
print(f"Orbital period in years: {T_halley*conversion_seconds_to_years}")
print(f"Velocity at aphelion: {v2_halley:.2f}m/s")
print(f"Distance at the sun in aphelion: {l2_halley*conversion_meters_to_kilometers:.2f}km")
print(f"Excentricity: {e_halley:.4f}")
```

## Exercise 2.7 — Catalan numbers

$C_0 = 1$, $C_{n+1} = \frac{4n+2}{n+2} C_n$. Print all Catalan numbers $\leq 10^9$ in increasing order.


```python
limit = 1_000_000_000
cn = 1
n = 0

while cn <= limit:
    print(cn)
    cn = ((4*n+2) * cn) // (n+2)
    n += 1
```

## Exercise 2.9 — Madelung constant

For NaCl on a cubic lattice:

$$M = \sum_{i,j,k=-L}^{L} \frac{(-1)^{i+j+k+1}}{\sqrt{i^2+j^2+k^2}} \quad (i,j,k) \neq (0,0,0)$$


```python
import numpy as np

# Charge of electron and term "4pie0a" cancels. Then M = sum_{i,j,k=-L to L, not i=j=k=0} [(-1)^(i+j+k+1)] / sqrt(i^2 + j^2 + k^2)
L = 100 # Size of the cubic box

i, j, k = np.ogrid[-L:L+1, -L:L+1, -L:L+1]
distance = np.sqrt(i**2 + j**2 + k**2)
distance[L, L, L] = np.inf
sign = np.where((i+j+k) % 2 == 0, -1.0, 1.0)
M = np.sum(sign / distance)
print(f"Madelung constant (L={L}): {M:.5f}")
```

## Exercise 2.10 — Semi-empirical mass formula

$$B = a_1 A - a_2 A^{2/3} - a_3 \frac{Z^2}{A^{1/3}} - a_4 \frac{(A-2Z)^2}{A} + \frac{a_5}{A^{1/2}}$$

Constants (MeV): $a_1=15.8$, $a_2=18.3$, $a_3=0.714$, $a_4=23.2$.

$a_5 = 0$ if $A$ odd; $12.0$ if $A,Z$ even; $-12.0$ if $A$ even, $Z$ odd.

- **(a–b)** Binding energy and per nucleon for given $A$, $Z$.
- **(c)** Most stable isotope for given $Z$ ($A$ from $Z$ to $3Z$).
- **(d)** Most stable $Z$ from 1 to 100.


```python
import numpy as np

a1 = 15.8
a2 = 18.3
a3 = 0.714
a4 = 23.2

# ==========================================
print("Item A and B")
A = int(input("Enter mass number A: "))
Z = int(input("Enter atomic number Z: "))

if A % 2 == 1:
    a5 = 0.0
else:
    if Z % 2 == 0:
        a5 = 12.0
    else:
        a5 = -12.0
        
B = a1*A - a2*A**(2/3) - a3*(Z**2)/(A**(1/3)) - a4*(A-2*Z)**2/A + a5/(A**(1/2))
print(f"Binding energy: {B:.2f} MeV")
print(f"Binding energy per nucleon: {B/A:.2f} MeV")

# ==========================================
print("\n---")
print("Item C")
Z = int(input("Enter atomic number Z: "))
A_values = np.arange(start=Z, stop=3*Z+1)

max_per_nucleon = 0.0
stable_A = 0

for A in A_values:
    if A % 2 == 1:
        a5 = 0.0
    else:
        if Z % 2 == 0:
            a5 = 12.0
        else:
            a5 = -12.0
            
    B = a1*A - a2*A**(2/3) - a3*(Z**2)/(A**(1/3)) - a4*(A-2*Z)**2/A + a5/(A**(1/2))
    b_per_nucleon = B / A
    
    if b_per_nucleon > max_per_nucleon:
        max_per_nucleon = b_per_nucleon
        stable_A = A

print(f"Stablest nucleus is: A: {stable_A}, Z: {Z}")
print(f"Binding energy per nucleon: {max_per_nucleon:.2f} MeV")

# ==========================================
print("\n---")
print("Item D")

z_values = np.arange(start=1, stop=101)
B_max_per_nucleon = np.empty(z_values.size)
A_most_stable = np.empty(z_values.size)

for index, z in enumerate(z_values):
    A_values = np.arange(start=z, stop=3*z+1)  
    max_per_nucleon = 0.0
    stable_A = 0
    
    for A in A_values:
        if A % 2 == 1:
            a5 = 0.0
        else:
            if z % 2 == 0: 
                a5 = 12.0
            else:
                a5 = -12.0
                
        B = a1*A - a2*A**(2/3) - a3*(z**2)/(A**(1/3)) - a4*(A-2*z)**2/A + a5/(A**(1/2)) 
        b_per_nucleon = B / A
        
        if b_per_nucleon > max_per_nucleon:
            max_per_nucleon = b_per_nucleon
            stable_A = A
            
    B_max_per_nucleon[index] = max_per_nucleon
    A_most_stable[index] = stable_A
    
idx = np.argmax(B_max_per_nucleon)
print(f"Z with maximum binding energy per nucleon is: {z_values[idx]} (Nickel)")
print(f"Its most stable mass number (A) is: {int(A_most_stable[idx])}")
print(f"Max binding energy per nucleon: {B_max_per_nucleon[idx]:.4f} MeV")
```

## Exercise 2.11 — Binomial coefficients

The binomial coefficient $\binom{n}{k}$ is given by:

$$\binom{n}{k} = \frac{n!}{k!(n-k)!} = \frac{n \times (n-1) \times \dots \times (n-k+1)}{1 \times 2 \times \dots \times k}$$

where $k \geq 1$, and $\binom{n}{0} = 1$.

---

- **(a)** Write a Python user-defined function `binomial(n, k)` that calculates the coefficient as an integer.
- **(b)** Print the first 20 lines of Pascal's triangle.
- **(c)** Calculate the probability of exactly 60 heads in 100 coin tosses, and the probability of 60 or more heads.

```python
def factorial(n):
    result = 1
    for i in range(2,n+1):
        result *= i
    return result

# Item a
def binomial(n,k):
    if k > n:
        return 0
    result = factorial(n) // (factorial(k)*factorial(n-k))
    return int(result)

# Item b
for i in range(1,21):
    line = [binomial(i,k) for k in range(0,i+1)]
    print(line)

# Item c
# Probability of exactly 60 heads in 100 coin tosses:
p = binomial(100, 60) * (1/2)**100
print(f"Probability to get 60 heads in 100 coin tosses: {p}")

# Probability of 60 or more heads:
pm = sum([binomial(100,k)*(1/2)**100 for k in range(60,101)])
print(f"Probability to get 60 heads or more in 100 coin tosses: {pm}")
```

## Exercise 2.12 — Prime numbers

The program in Example 2.8 is not a very efficient way of calculating prime numbers: it checks each number to see if it is divisible by any number less than it. We can develop a much faster program for prime numbers by making use of the following observations:

* **a)** A number $n$ is prime if it has no prime factors less than $n$. Hence we only need to check if it is divisible by other primes.
* **b)** If a number $n$ is non-prime, having a factor $r$, then $n = rs$, where $s$ is also a factor. If $r \geq \sqrt{n}$ then $n = rs \geq \sqrt{n}s$, which implies that $s \leq \sqrt{n}$. In other words, any non-prime must have factors, and hence also prime factors, less than or equal to $\sqrt{n}$. Thus to determine if a number is prime we have to check its prime factors only up to and including $\sqrt{n}$—if there are none then the number is prime.
* **c)** If we find even a single prime factor less than $\sqrt{n}$ then we know that the number is non-prime, and hence there is no need to check any further—we can abandon this number and move on to something else.

---

Write a Python program that finds all the primes up to ten thousand. Create a list to store the primes, which starts out with just the one prime number 2 in it. Then for each number $n$ from 3 to 10 000 check whether the number is divisible by any of the primes in the list up to and including $\sqrt{n}$. As soon as you find a single prime factor you can stop checking the rest of them—you know $n$ is not a prime. If you find no prime factors $\sqrt{n}$ or less then $n$ is prime and you should add it to the list. You can print out the list all in one go at the end of the program, or you can print out the individual numbers as you find them.

```python
list_primes = [2]

for n in range(3,10001):
    limit = int(n**(1/2))
    for p in list_primes:
        if n % p == 0:
            break
        if p > limit:
            list_primes.append(n)
            break

print(list_primes)
```
