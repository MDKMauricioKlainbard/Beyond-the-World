## Exercise 2A.1

**Statement.**

Find a list of four distinct vectors in $\mathbf{F}^3$ whose span equals
$$\{(x, y, z) \in \mathbf{F}^3 : x + y + z = 0\}.$$

### Solution

Let $U = \{(x, y, z) \in \mathbf{F}^3 : x + y + z = 0\}$. We choose the following list of four distinct vectors:
$$v_1 = (1, -2, 1), \quad v_2 = (2, -5, 3), \quad v_3 = (2, -4, 2), \quad v_4 = (0, 0, 0)$$

To prove that $\text{span}(v_1, v_2, v_3, v_4) = U$ without using the concept of dimension, we will show double containment: $\text{span}(v_1, v_2, v_3, v_4) \subseteq U$ and $U \subseteq \text{span}(v_1, v_2, v_3, v_4)$.

**Forward Containment ($\text{span}(v_1, v_2, v_3, v_4) \subseteq U$):**
We verify that each vector satisfies the defining condition of $U$ by adding its components:
- For $v_1$: $1 + (-2) + 1 = 0 \implies v_1 \in U$
- For $v_2$: $2 + (-5) + 3 = 0 \implies v_2 \in U$
- For $v_3$: $2 + (-4) + 2 = 0 \implies v_3 \in U$
- For $v_4$: $0 + 0 + 0 = 0 \implies v_4 \in U$

Since $U$ is a subspace of $\mathbf{F}^3$ and contains all four vectors, any linear combination of these vectors must also belong to $U$. Therefore, $\text{span}(v_1, v_2, v_3, v_4) \subseteq U$.

**Backward Containment ($U \subseteq \text{span}(v_1, v_2, v_3, v_4)$):**
Let $(x, y, z)$ be an arbitrary vector in $U$. The condition $x + y + z = 0$ implies that $z = -x - y$. Thus, any vector in $U$ is uniquely characterized by its first two coordinates and takes the form $(x, y, -x - y)$.

We must show that there exist scalars $c_1, c_2, c_3, c_4 \in \mathbf{F}$ such that:
$$(x, y, -x - y) = c_1 v_1 + c_2 v_2 + c_3 v_3 + c_4 v_4$$

We can simplify our search by setting $c_3 = 0$ and $c_4 = 0$, attempting to write the vector solely as a combination of $v_1$ and $v_2$:
$$(x, y, -x - y) = c_1(1, -2, 1) + c_2(2, -5, 3)$$
$$(x, y, -x - y) = (c_1 + 2c_2, -2c_1 - 5c_2, c_1 + 3c_2)$$

This yields the following system of linear equations for the first two components:
1) $c_1 + 2c_2 = x$
2) $-2c_1 - 5c_2 = y$

From equation (1), we isolate $c_1 = x - 2c_2$. Substituting this into equation (2) gives:
$$-2(x - 2c_2) - 5c_2 = y \implies -2x + 4c_2 - 5c_2 = y \implies c_2 = -2x - y$$

Substituting $c_2$ back into the expression for $c_1$:
$$c_1 = x - 2(-2x - y) = x + 4x + 2y = 5x + 2y$$

Now, we check if these chosen scalars satisfy the equation for the third component:
$$c_1 + 3c_2 = (5x + 2y) + 3(-2x - y) = 5x + 2y - 6x - 3y = -x - y$$

The third component matches perfectly. This means that for any vector $(x, y, z) \in U$, we can explicitly express it as:
$$(x, y, z) = (5x + 2y)v_1 + (-2x - y)v_2 + 0v_3 + 0v_4$$

This confirms that $(x, y, z) \in \text{span}(v_1, v_2, v_3, v_4)$, establishing that $U \subseteq \text{span}(v_1, v_2, v_3, v_4)$.

Since both containments hold, we conclude that $\text{span}(v_1, v_2, v_3, v_4) = U \quad \blacksquare$

## Exercise 2.A.5

**Statement.**

Find a number $t$ such that  

$$
(3, 1, 4),\quad (2, -3, 5),\quad (5, 9, t)
$$

is not linearly independent in $\mathbb{R}^3$.


### Solution

For the three vectors to be linearly dependent, the third must lie in the span of the first two (since the first two are not multiples). We seek scalars $a$ and $b$ satisfying  

$$
(5, 9, t) = a(3, 1, 4) + b(2, -3, 5).
$$

This gives the system  

$$
3a + 2b = 5
$$

$$
a - 3b = 9
$$

$$
4a + 5b = t
$$

From the second equation, $a = 9 + 3b$. Substitute into the first:

$$
3(9 + 3b) + 2b = 5
$$

$$
27 + 9b + 2b = 5
$$

$$
27 + 11b = 5
$$

$$
11b = -22
$$

$$
b = -2
$$

Then $a = 9 + 3(-2) = 9 - 6 = 3$. Now compute $t$:

$$
t = 4a + 5b = 4 \cdot 3 + 5 \cdot (-2) = 12 - 10 = 2
$$

Thus for $t = 2$ the third vector is a combination of the first two, so the three vectors are linearly dependent (i.e., not linearly independent).  

Answer: $t = 2$. ∎

## Exercise 2.A.19

**Statement.**

Prove that the real vector space of all continuous real‑valued functions on the interval $[0, 1]$ is infinite‑dimensional.


### Solution

Let $V$ be the space of all continuous functions $f: [0,1] \to \mathbb{R}$.  
Consider the set of monomials $\{1, x, x^2, x^3, \dots\}$. Each monomial $x^n$ is continuous, so it belongs to $V$.

These monomials are linearly independent: if a finite linear combination $a_0 + a_1 x + \dots + a_n x^n = 0$ (as a function on $[0,1]$), then the polynomial is identically zero, hence all coefficients $a_i = 0$.

Thus $V$ contains an infinite linearly independent set. In a finite‑dimensional vector space, every linearly independent set has size at most the dimension. Because we have an arbitrarily large finite linearly independent set (e.g. $\{1, x, \dots, x^m\}$ for any $m$), the dimension cannot be finite. Therefore $V$ is infinite‑dimensional.

∎
