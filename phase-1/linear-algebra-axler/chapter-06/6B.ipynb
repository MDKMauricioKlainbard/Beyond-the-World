## Exercise 6B.8

**Statement.**

Make $\mathcal{P}_2(\mathbf{R})$ into an inner product space by defining $\langle p, q \rangle = \int_0^1 p(x)q(x) \, dx$ for all $p, q \in \mathcal{P}_2(\mathbf{R})$.

(a) Apply the Gram–Schmidt procedure to the basis $1, x, x^2$ to produce an orthonormal basis of $\mathcal{P}_2(\mathbf{R})$.
(b) The differentiation operator (the operator that takes $p$ to $p'$) on $\mathcal{P}_2(\mathbf{R})$ has an upper-triangular matrix with respect to the basis $1, x, x^2$, which is not an orthonormal basis. Find the matrix of the differentiation operator on $\mathcal{P}_2(\mathbf{R})$ with respect to the orthonormal basis produced in (a) and verify that this matrix is upper triangular, as expected from the proof of 6.37.

### Solution

**(a)** Let $v_1 = 1$, $v_2 = x$, and $v_3 = x^2$. We apply the Gram-Schmidt procedure to construct an orthonormal basis $(e_1, e_2, e_3)$.

1. First vector:
   $$u_1 = v_1 = 1$$
   $$\|u_1\|^2 = \int_0^1 1 \, dx = 1 \implies e_1 = 1$$

2. Second vector:
   $$\langle v_2, e_1 \rangle = \int_0^1 x(1) \, dx = \frac{1}{2}$$
   $$u_2 = v_2 - \langle v_2, e_1 \rangle e_1 = x - \frac{1}{2}$$
   $$\|u_2\|^2 = \int_0^1 \left(x - \frac{1}{2}\right)^2 dx = \int_0^1 \left(x^2 - x + \frac{1}{4}\right) dx = \frac{1}{3} - \frac{1}{2} + \frac{1}{4} = \frac{1}{12}$$
   $$\|u_2\| = \frac{1}{\sqrt{12}} = \frac{1}{2\sqrt{3}} \implies e_2 = \sqrt{3}(2x - 1)$$

3. Third vector:
   $$\langle v_3, e_1 \rangle = \int_0^1 x^2(1) \, dx = \frac{1}{3}$$
   $$\langle v_3, e_2 \rangle = \int_0^1 x^2 \sqrt{3}(2x - 1) \, dx = \sqrt{3} \int_0^1 (2x^3 - x^2) \, dx = \sqrt{3}\left(\frac{1}{2} - \frac{1}{3}\right) = \frac{\sqrt{3}}{6} = \frac{1}{2\sqrt{3}}$$
   $$u_3 = v_3 - \langle v_3, e_1 \rangle e_1 - \langle v_3, e_2 \rangle e_2 = x^2 - \frac{1}{3} - \frac{1}{2\sqrt{3}}\sqrt{3}(2x - 1) = x^2 - x + \frac{1}{6}$$
   $$\|u_3\|^2 = \int_0^1 \left(x^2 - x + \frac{1}{6}\right)^2 dx = \frac{1}{180} \implies \|u_3\| = \frac{1}{6\sqrt{5}}$$
   $$e_3 = \sqrt{5}(6x^2 - 6x + 1)$$

Thus, the orthonormal basis is:
$$e_1 = 1, \quad e_2 = \sqrt{3}(2x - 1), \quad e_3 = \sqrt{5}(6x^2 - 6x + 1)$$

**(b)** Let $D \in \mathcal{L}(\mathcal{P}_2(\mathbf{R}))$ be the differentiation operator $Dp = p'$. We compute the action of $D$ on each vector of our orthonormal basis:

1. For $e_1$:
   $$D(e_1) = \frac{d}{dx}(1) = 0 = 0e_1 + 0e_2 + 0e_3$$

2. For $e_2$:
   $$D(e_2) = \frac{d}{dx}(\sqrt{3}(2x - 1)) = 2\sqrt{3} = 2\sqrt{3}e_1 + 0e_2 + 0e_3$$

3. For $e_3$:
   $$D(e_3) = \frac{d}{dx}(\sqrt{5}(6x^2 - 6x + 1)) = \sqrt{5}(12x - 6) = 6\sqrt{5}(2x - 1)$$
   Substituting $2x - 1 = \frac{e_2}{\sqrt{3}}$ yields:
   $$D(e_3) = 6\sqrt{5}\left(\frac{e_2}{\sqrt{3}}\right) = 2\sqrt{15}e_2 = 0e_1 + 2\sqrt{15}e_2 + 0e_3$$

Gathering the coordinate vectors as columns, the matrix representation $\mathcal{M}(D)$ with respect to $(e_1, e_2, e_3)$ is:
$$
\mathcal{M}(D) = \begin{pmatrix}
0 & 2\sqrt{3} & 0 \\
0 & 0 & 2\sqrt{15} \\
0 & 0 & 0
\end{pmatrix}
$$

Since all entries on and below the main diagonal are zero, the matrix is strictly upper triangular. ∎
