## Exercise 6A.1

**Statement.**

Prove or give a counterexample: If $v_1, \dots, v_m \in V$, then
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

### Solution

The statement is true. We provide a proof using the fundamental positivity property of inner product spaces.

Let $w \in V$ be the vector defined by the sum of all elements in the collection:
$$
w = \sum_{i=1}^m v_i
$$

By applying the additivity and linearity properties of the inner product to both arguments, expanding the inner product of $w$ with itself yields the original double summation:
$$
\langle w, w \rangle = \left\langle \sum_{j=1}^m v_j, \sum_{k=1}^m v_k \right\rangle = \sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle
$$

By the fundamental positivity property of inner products, the inner product of any vector with itself is always non-negative:
$$
\langle w, w \rangle \ge 0
$$

Therefore, it immediately follows that:
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

This completes the proof. ∎
