## Exercise 6A.1

**Statement.**

Prove or give a counterexample: If $v_1, \dots, v_m \in V$, then
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

### Solution

The statement is true. We provide a proof using the fundamental positivity property of inner product spaces.

Let $w \in V$ be the vector defined by the sum of all elements in the collection:
$$
w = \sum_{i=1}^m v_i
$$

By applying the additivity and linearity properties of the inner product to both arguments, expanding the inner product of $w$ with itself yields the original double summation:
$$
\langle w, w \rangle = \left\langle \sum_{j=1}^m v_j, \sum_{k=1}^m v_k \right\rangle = \sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle
$$

By the fundamental positivity property of inner products, the inner product of any vector with itself is always non-negative:
$$
\langle w, w \rangle \ge 0
$$

Therefore, it immediately follows that:
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

This completes the proof. ∎

## Exercise 6A.11

**Statement.**

Find vectors $u, v \in \mathbf{R}^2$ such that $u$ is a scalar multiple of $(1,3)$, $v$ is orthogonal to $(1,3)$, and $(1,2) = u + v$.

### Solution

We solve this by setting up a system of linear equations based on the given geometric conditions.

Let $u$ be a scalar multiple of $(1,3)$ for some scalar $c \in \mathbf{R}$:
$$
u = c(1,3) = (c, 3c)
$$

Let $v = (x,y) \in \mathbf{R}^2$. Since $v$ is orthogonal to $(1,3)$, their standard inner product must equal zero:
$$
\langle v, (1,3) \rangle = x + 3y = 0
$$

We are given that $u + v = (1,2)$, which relates the components as follows:
$$
(1,2) = (c + x, 3c + y)
$$

This allows us to express $x$ and $y$ in terms of $c$:
$$
\begin{aligned}
x &= 1 - c \\
y &= 2 - 3c
\end{aligned}
$$

Substituting these expressions into the orthogonality condition $x + 3y = 0$:
$$
(1 - c) + 3(2 - 3c) = 0
$$
$$
1 - c + 6 - 9c = 0
$$
$$
7 - 10c = 0 \implies c = \frac{7}{10}
$$

Now, we substitute $c = \frac{7}{10}$ back to determine the explicit vectors $u$ and $v$:
$$
u = \frac{7}{10}(1,3) = \left(\frac{7}{10}, \frac{21}{10}\right)
$$
$$
x = 1 - \frac{7}{10} = \frac{3}{10}
$$
$$
y = 2 - 3\left(\frac{7}{10}\right) = -\frac{1}{10}
$$

Thus, the unique pair of vectors satisfying the conditions is:
$$
u = \left(\frac{7}{10}, \frac{21}{10}\right), \quad v = \left(\frac{3}{10}, -\frac{1}{10}\right)
$$

We verify that $u + v = (1,2)$ and $\langle u, v \rangle = \frac{21}{100} - \frac{21}{100} = 0$. ∎

## Exercise 6A.16

**Statement.**

The angle between two vectors (thought of as arrows with initial point at the origin) in $\mathbf{R}^2$ or $\mathbf{R}^3$ can be defined geometrically. However, geometry is not as clear in $\mathbf{R}^n$ for $n > 3$. Thus the angle between two nonzero vectors $x, y \in \mathbf{R}^n$ is defined to be
$$
\arccos \frac{\langle x, y \rangle}{\|x\| \|y\|},
$$
where the motivation for this definition comes from Exercise 15. Explain why the Cauchy–Schwarz inequality is needed to show that this definition makes sense.

### Solution

For the definition of the angle to make sense, the argument of the inverse cosine function must fall within its valid domain. The domain of the real-valued function $f(t) = \arccos t$ is the closed interval $[-1, 1]$. Therefore, we must guarantee that for any nonzero vectors $x, y \in \mathbf{R}^n$:
$$
-1 \le \frac{\langle x, y \rangle}{\|x\| \|y\|} \le 1
$$

This condition can be compactly rewritten by taking the absolute value of the expression:
$$
\frac{|\langle x, y \rangle|}{\|x\| \|y\|} \le 1
$$

Since $x$ and $y$ are nonzero vectors, their norms $\|x\|$ and $\|y\|$ are strictly positive. Multiplying both sides of the inequality by the product of the norms yields the equivalent condition:
$$
|\langle x, y \rangle| \le \|x\| \|y\|
$$

The Cauchy–Schwarz inequality states precisely that $|\langle x, y \rangle| \le \|x\| \|y\|$ holds true for all vectors in any inner product space. Consequently, dividing back by the strictly positive product $\|x\| \|y\|$ guarantees that the ratio always falls within $[-1, 1]$. 

Without the Cauchy–Schwarz inequality, the quotient could potentially yield values outside the interval $[-1, 1]$, rendering the arccosine function undefined and the geometric notion of an angle in higher dimensions invalid. ∎

## Exercise 6A.21

**Statement.**

Suppose $u, v \in V$ are such that
$$
\|u\| = 3, \quad \|u + v\| = 4, \quad \|u - v\| = 6
$$

What number does $\|v\|$ equal?

### Solution

We can determine the value of $\|v\|$ by applying the parallelogram law, which states that for any vectors $u, v$ in an inner product space:
$$
\|u + v\|^2 + \|u - v\|^2 = 2(\|u\|^2 + \|v\|^2)
$$

Substituting the given values into the equation yields:
$$
4^2 + 6^2 = 2(3^2 + \|v\|^2)
$$
$$
16 + 36 = 2(9 + \|v\|^2)
$$
$$
52 = 18 + 2\|v\|^2
$$

Isolating $\|v\|^2$:
$$
2\|v\|^2 = 34 \implies \|v\|^2 = 17
$$

Since the norm of a vector is always non-negative, we take the positive square root:
$$
\|v\| = \sqrt{17}
$$

Thus, $\|v\|$ equals $\sqrt{17}$. ∎

## Exercise 6A.26

**Statement.**

Suppose $V$ is a real inner product space. Prove that
$$
\langle u, v \rangle = \frac{\|u + v\|^2 - \|u - v\|^2}{4}
$$
for all $u, v \in V$.

### Solution

The statement is true. We provide a proof by expanding the norms using the definition and properties of the inner product on a real vector space.

First, we expand the term $\|u + v\|^2$ using the additivity and homogeneity of the inner product:
$$
\|u + v\|^2 = \langle u + v, u + v \rangle = \langle u, u \rangle + \langle u, v \rangle + \langle v, u \rangle + \langle v, v \rangle
$$

Since $V$ is a real inner product space, the inner product is symmetric, which implies $\langle u, v \rangle = \langle v, u \rangle$. Thus, the expression simplifies to:
$$
\|u + v\|^2 = \|u\|^2 + 2\langle u, v \rangle + \|v\|^2
$$

Next, we expand the term $\|u - v\|^2$ in an identical manner:
$$
\|u - v\|^2 = \langle u - v, u - v \rangle = \langle u, u \rangle - \langle u, v \rangle - \langle v, u \rangle + \langle v, v \rangle
$$

Applying the symmetry condition $\langle u, v \rangle = \langle v, u \rangle$ again, we obtain:
$$
\|u - v\|^2 = \|u\|^2 - 2\langle u, v \rangle + \|v\|^2
$$

Now, we subtract the second equation from the first:
$$
\|u + v\|^2 - \|u - v\|^2 = \left( \|u\|^2 + 2\langle u, v \rangle + \|v\|^2 \right) - \left( \|u\|^2 - 2\langle u, v \rangle + \|v\|^2 \right)
$$

Simplifying the right-hand side by canceling the common terms $\|u\|^2$ and $\|v\|^2$ yields:
$$
\|u + v\|^2 - \|u - v\|^2 = 4\langle u, v \rangle
$$

Finally, dividing both sides by $4$, we arrive at the desired identity:
$$
\langle u, v \rangle = \frac{\|u + v\|^2 - \|u - v\|^2}{4}
$$

This completes the proof. ∎
