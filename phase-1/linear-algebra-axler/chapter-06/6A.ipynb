## Exercise 6A.1

**Statement.**

Prove or give a counterexample: If $v_1, \dots, v_m \in V$, then
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

### Solution

The statement is true. We provide a proof using the fundamental positivity property of inner product spaces.

Let $w \in V$ be the vector defined by the sum of all elements in the collection:
$$
w = \sum_{i=1}^m v_i
$$

By applying the additivity and linearity properties of the inner product to both arguments, expanding the inner product of $w$ with itself yields the original double summation:
$$
\langle w, w \rangle = \left\langle \sum_{j=1}^m v_j, \sum_{k=1}^m v_k \right\rangle = \sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle
$$

By the fundamental positivity property of inner products, the inner product of any vector with itself is always non-negative:
$$
\langle w, w \rangle \ge 0
$$

Therefore, it immediately follows that:
$$
\sum_{j=1}^m \sum_{k=1}^m \langle v_j, v_k \rangle \ge 0
$$

This completes the proof. ∎

## Exercise 6A.21

**Statement.**

Suppose $u, v \in V$ are such that
$$
\|u\| = 3, \quad \|u + v\| = 4, \quad \|u - v\| = 6
$$

What number does $\|v\|$ equal?

### Solution

We can determine the value of $\|v\|$ by applying the parallelogram law, which states that for any vectors $u, v$ in an inner product space:
$$
\|u + v\|^2 + \|u - v\|^2 = 2(\|u\|^2 + \|v\|^2)
$$

Substituting the given values into the equation yields:
$$
4^2 + 6^2 = 2(3^2 + \|v\|^2)
$$
$$
16 + 36 = 2(9 + \|v\|^2)
$$
$$
52 = 18 + 2\|v\|^2
$$

Isolating $\|v\|^2$:
$$
2\|v\|^2 = 34 \implies \|v\|^2 = 17
$$

Since the norm of a vector is always non-negative, we take the positive square root:
$$
\|v\| = \sqrt{17}
$$

Thus, $\|v\|$ equals $\sqrt{17}$. ∎

## Exercise 6A.26

**Statement.**

Suppose $V$ is a real inner product space. Prove that
$$
\langle u, v \rangle = \frac{\|u + v\|^2 - \|u - v\|^2}{4}
$$
for all $u, v \in V$.

### Solution

The statement is true. We provide a proof by expanding the norms using the definition and properties of the inner product on a real vector space.

First, we expand the term $\|u + v\|^2$ using the additivity and homogeneity of the inner product:
$$
\|u + v\|^2 = \langle u + v, u + v \rangle = \langle u, u \rangle + \langle u, v \rangle + \langle v, u \rangle + \langle v, v \rangle
$$

Since $V$ is a real inner product space, the inner product is symmetric, which implies $\langle u, v \rangle = \langle v, u \rangle$. Thus, the expression simplifies to:
$$
\|u + v\|^2 = \|u\|^2 + 2\langle u, v \rangle + \|v\|^2
$$

Next, we expand the term $\|u - v\|^2$ in an identical manner:
$$
\|u - v\|^2 = \langle u - v, u - v \rangle = \langle u, u \rangle - \langle u, v \rangle - \langle v, u \rangle + \langle v, v \rangle
$$

Applying the symmetry condition $\langle u, v \rangle = \langle v, u \rangle$ again, we obtain:
$$
\|u - v\|^2 = \|u\|^2 - 2\langle u, v \rangle + \|v\|^2
$$

Now, we subtract the second equation from the first:
$$
\|u + v\|^2 - \|u - v\|^2 = \left( \|u\|^2 + 2\langle u, v \rangle + \|v\|^2 \right) - \left( \|u\|^2 - 2\langle u, v \rangle + \|v\|^2 \right)
$$

Simplifying the right-hand side by canceling the common terms $\|u\|^2$ and $\|v\|^2$ yields:
$$
\|u + v\|^2 - \|u - v\|^2 = 4\langle u, v \rangle
$$

Finally, dividing both sides by $4$, we arrive at the desired identity:
$$
\langle u, v \rangle = \frac{\|u + v\|^2 - \|u - v\|^2}{4}
$$

This completes the proof. ∎
