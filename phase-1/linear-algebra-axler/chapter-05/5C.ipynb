# Section 5C – Upper-Triangular Matrices

*From Axler, 4th ed.*

### Exercise 5C.1

**Statement:**  
Prove or give a counterexample: If $T \in \mathcal{L}(V)$ and $T^2$ has an upper‑triangular matrix with respect to some basis of $V$, then $T$ has an upper‑triangular matrix with respect to some basis of $V$.

**Solution (counterexample over $\mathbb{R}$).**

Take $V = \mathbb{R}^2$ and define $T: \mathbb{R}^2 \to \mathbb{R}^2$ by  

$$
T(x, y) = (-y, x).
$$

In the standard basis, the matrix of $T$ is  

$$
\begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}.
$$

Then  

$$
T^2(x, y) = T(-y, x) = (-x, -y) = - (x, y),
$$

so $T^2 = -I$. The matrix of $T^2$ in the standard basis is $-I$, which is upper‑triangular.

Now, does $T$ have an upper‑triangular matrix with respect to some basis of $\mathbb{R}^2$?  
If such a basis existed, then the diagonal entries of that matrix would be the eigenvalues of $T$ (over $\mathbb{R}$). However, $T$ has no real eigenvalues because its minimal polynomial is $p(z) = z^2+1$ ($z^2+1=0$ has no real solutions). Consequently, $T$ cannot be represented by an upper‑triangular matrix over $\mathbb{R}$ (any upper‑triangular real matrix has all its eigenvalues real, appearing on the diagonal).  

Thus $T$ is a counterexample: $T^2$ is upper‑triangular (in fact diagonal), but $T$ itself is not upper‑triangular for any basis of $\mathbb{R}^2$.

∎

### Exercise 5C.2

**Statement:**  
Suppose $A$ and $B$ are upper‑triangular matrices of the same size, with $\alpha_1, \dots, \alpha_n$ on the diagonal of $A$ and $\beta_1, \dots, \beta_n$ on the diagonal of $B$.

(a) Show that $A + B$ is an upper‑triangular matrix with $\alpha_1 + \beta_1, \dots, \alpha_n + \beta_n$ on the diagonal.

(b) Show that $AB$ is an upper‑triangular matrix with $\alpha_1 \beta_1, \dots, \alpha_n \beta_n$ on the diagonal.

---

**Solution.**

#### (a) Sum of two upper‑triangular matrices

Recall the definition of an upper‑triangular matrix: $A_{i,j} = 0$ and $B_{i,j} = 0$ if and only if $i > j$.  
For $C = A + B$, we have $C_{i,j} = A_{i,j} + B_{i,j}$. If $i > j$, then both $A_{i,j} = 0$ and $B_{i,j} = 0$, so $C_{i,j} = 0$. Therefore $A + B$ is upper‑triangular.  

The diagonal entries are  

$$
C_{i,i} = A_{i,i} + B_{i,i} = \alpha_i + \beta_i.
$$

Thus the diagonal of $A+B$ consists of $\alpha_1+\beta_1, \dots, \alpha_n+\beta_n$.

---

#### (b) Product of two upper‑triangular matrices

For an entry $(i,j)$ with $i > j$, we consider the sum  

$$
(AB)_{i,j} = \sum_{k=1}^n A_{i,k} B_{k,j}.
$$

We split the sum into two parts: $k < i$ and $k \ge i$.

- If $k < i$, then because $A$ is upper‑triangular, $A_{i,k} = 0$.  
- If $k \ge i$, then $k \ge i > j$, so $k > j$. Because $B$ is upper‑triangular, $B_{k,j} = 0$.

Thus every term $A_{i,k} B_{k,j} = 0$, and consequently $(AB)_{i,j} = 0$ for all $i > j$. Therefore $AB$ is upper‑triangular.

For a diagonal entry $(i,i)$ we have  

$$
(AB)_{i,i} = \sum_{k=1}^n A_{i,k} B_{k,i}.
$$

Since $A$ is upper‑triangular, $A_{i,k} = 0$ when $i > k$.  
Since $B$ is upper‑triangular, $B_{k,i} = 0$ when $k > i$.

Hence the only possible non‑zero term occurs when $k = i$:

$$
(AB)_{i,i} = A_{i,i} B_{i,i} = \alpha_i \beta_i.
$$

Thus the diagonal of $AB$ consists of the products $\alpha_1 \beta_1, \dots, \alpha_n \beta_n$.

∎

### Exercise 5C.4

**Statement:**  
Give an example of an operator whose matrix with respect to some basis contains only $0$’s on the diagonal, but the operator is invertible.

*This exercise shows that the result about upper‑triangular matrices and invertibility (5.41) fails without the hypothesis that the matrix is upper‑triangular.*

**Solution.**

Consider the operator $T : \mathbb{R}^2 \to \mathbb{R}^2$ defined by  

$$
T(x, y) = (y, x).
$$

Take the standard basis $\{e_1 = (1,0), e_2 = (0,1)\}$. Then  

$$
T(e_1) = (0,1) = 0 \cdot e_1 + 1 \cdot e_2,
$$
$$
T(e_2) = (1,0) = 1 \cdot e_1 + 0 \cdot e_2.
$$

Hence the matrix of $T$ with respect to $\{e_1, e_2\}$ is  

$$
\mathcal{M}(T) = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix},
$$

which has zeros on the diagonal.

We can prove that $T$ is injective: $T(x,y) = (0,0) = (y,x)$ implies $x = y = 0$. Hence $T$ is invertible.

Thus $T$ is an invertible operator whose matrix (in the standard basis) has only zeros on the diagonal.

∎

### Exercise 5C.11

**Statement:**  
Suppose $\mathbf{F} = \mathbb{C}$ and $V$ is finite‑dimensional. Prove that if $T \in \mathcal{L}(V)$, then there exists a basis of $V$ with respect to which $T$ has a lower‑triangular matrix.

**Proof.**

We use the dual space $V' = \mathcal{L}(V, \mathbb{C})$.  
The dual map $T' \in \mathcal{L}(V')$ is defined by $T'(\varphi) = \varphi \circ T$ for every $\varphi \in V'$.

Because $V$ is finite‑dimensional over $\mathbb{C}$, $V'$ is also finite‑dimensional.  
A standard theorem states that **every operator on a finite‑dimensional complex vector space has an upper‑triangular matrix** with respect to some basis.  
Apply this theorem to $T'$. Hence there exists a basis $\{\varphi_1, \dots, \varphi_n\}$ of $V'$ such that the matrix of $T'$ with respect to $\{\varphi_1, \dots, \varphi_n\}$ is upper‑triangular.  
Let $A = (a_{ij})$ be that matrix; i.e.

$$
T'(\varphi_j) = \sum_{i=1}^j a_{ij} \varphi_i \qquad (1 \le j \le n),
$$

where $a_{ij} = 0$ whenever $i > j$.

Now let $\{v_1, \dots, v_n\}$ be the **dual basis** of $V$ corresponding to $\{\varphi_1, \dots, \varphi_n\}$; that is,

$$
\varphi_i(v_j) = \delta_{ij} = \begin{cases}
1 & \text{if } i=j,\\
0 & \text{if } i \neq j.
\end{cases}
$$

We claim that the matrix of $T$ with respect to $\{v_1, \dots, v_n\}$ is the transpose $A^{\mathsf{T}}$, which is lower‑triangular.

Compute the $(i,j)$-entry of the matrix of $T$. By definition, $(B_{ij})$ satisfies

$$
T(v_j) = \sum_{i=1}^n B_{ij} v_i.
$$

Applying $\varphi_i$ to both sides gives

$$
\varphi_i(T(v_j)) = \sum_{k=1}^n B_{kj} \varphi_i(v_k) = B_{ij}.
$$

On the other hand, using the definition of $T'$,

$$
\varphi_i(T(v_j)) = (T'(\varphi_i))(v_j).
$$

Write $T'(\varphi_i)$ as a linear combination of the $\varphi$’s:

$$
T'(\varphi_i) = \sum_{k=1}^i a_{ki} \varphi_k \qquad (\text{because } a_{ki}=0 \text{ for } k > i).
$$

Then

$$
\varphi_i(T(v_j)) = \sum_{k=1}^i a_{ki} \varphi_k(v_j) = \sum_{k=1}^i a_{ki} \delta_{kj}.
$$

The only term that survives is when $k = j$. Since $j \le i$ (otherwise the term is absent because $k$ runs only up to $i$), we obtain

$$
\varphi_i(T(v_j)) = 
\begin{cases}
a_{ji} & \text{if } j \le i,\\
0 & \text{if } j > i.
\end{cases}
$$

Thus $B_{ij} = a_{ji}$ for all $i,j$. Therefore the matrix $B = (B_{ij})$ is exactly the **transpose** of $A$. Because $A$ is upper‑triangular, $A_{ij}=0$ when $i > j$. Hence $B_{ij}=0$ when $i < j$, which means $B$ is **lower‑triangular**.

Consequently, with respect to the basis $\{v_1, \dots, v_n\}$ of $V$, the matrix of $T$ is lower‑triangular. ∎
