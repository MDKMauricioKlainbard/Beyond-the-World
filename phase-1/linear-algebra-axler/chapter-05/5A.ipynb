## Exercise 5.A.21

**Statement.**

Suppose $T \in \mathcal{L}(V)$ is invertible.

(a) Suppose $\lambda \in \mathbb{F}$ with $\lambda \neq 0$. Prove that $\lambda$ is an eigenvalue of $T$ if and only if $\frac{1}{\lambda}$ is an eigenvalue of $T^{-1}$.

(b) Prove that $T$ and $T^{-1}$ have the same eigenvectors.


### Solution

#### Part (a)

**($\Rightarrow$)**  
Assume $\lambda$ is an eigenvalue of $T$. Then there exists a non‑zero vector $v \in V$ such that $Tv = \lambda v$. Because $T$ is invertible, $\lambda \neq 0$ (otherwise $T$ would not be injective). Apply $T^{-1}$ to both sides:

$$
T^{-1}(Tv) = T^{-1}(\lambda v) \;\Longrightarrow\; v = \lambda T^{-1}v.
$$

Since $\lambda \neq 0$, we can divide by $\lambda$:

$$
T^{-1}v = \frac{1}{\lambda}\, v.
$$

Thus $v$ is an eigenvector of $T^{-1}$ with eigenvalue $1/\lambda$. Hence $1/\lambda$ is an eigenvalue of $T^{-1}$.

**($\Leftarrow$)**  
Assume $1/\lambda$ is an eigenvalue of $T^{-1}$. Then there exists a non‑zero $v \in V$ such that $T^{-1}v = \frac{1}{\lambda} v$. Apply $T$ to both sides:

$$
T(T^{-1}v) = T\!\left(\frac{1}{\lambda} v\right) \;\Longrightarrow\; v = \frac{1}{\lambda} Tv.
$$

Multiplying by $\lambda$ gives $Tv = \lambda v$. Therefore $\lambda$ is an eigenvalue of $T$.

---

#### Part (b)

Let $v$ be an eigenvector of $T$. Then $Tv = \lambda v$ for some $\lambda \neq 0$ (since $T$ is invertible). From part (a), we already saw that $T^{-1}v = \frac{1}{\lambda} v$, so $v$ is also an eigenvector of $T^{-1}$.

Conversely, let $v$ be an eigenvector of $T^{-1}$. Then $T^{-1}v = \mu v$ for some $\mu \neq 0$ (because $T^{-1}$ is also invertible). Applying $T$ to both sides yields $v = \mu Tv$, hence $Tv = \frac{1}{\mu} v$. Thus $v$ is an eigenvector of $T$.

Therefore $T$ and $T^{-1}$ have exactly the same set of eigenvectors. ∎

## Exercise 5.A.35

**Statement.**

Suppose $\lambda_1,\dots,\lambda_n$ is a list of distinct real numbers. Prove that the list $e^{\lambda_1 x},\dots,e^{\lambda_n x}$ is linearly independent in the vector space of real‑valued functions on $\mathbb{R}$.

*Hint: Let $V = \operatorname{span}(e^{\lambda_1 x},\dots,e^{\lambda_n x})$, and define an operator $D \in \mathcal{L}(V)$ by $Df = f'$. Find eigenvalues and eigenvectors of $D$.*


### Solution

Let $V = \operatorname{span}\{e^{\lambda_1 x},\dots,e^{\lambda_n x}\}$. Define $D : V \to V$ by $Df = f'$ (the derivative). First we check that $D$ indeed maps $V$ into $V$: for any linear combination $f = \sum_{i=1}^n c_i e^{\lambda_i x}$, we have  

$$
f' = \sum_{i=1}^n c_i \lambda_i e^{\lambda_i x} \in V.
$$

Now compute the action of $D$ on each basis function:

$$
D(e^{\lambda_i x}) = \lambda_i e^{\lambda_i x}.
$$

Thus each $e^{\lambda_i x}$ is an eigenvector of $D$ with eigenvalue $\lambda_i$. Because the eigenvalues $\lambda_1,\dots,\lambda_n$ are distinct, the corresponding eigenvectors are linearly independent (a standard result: eigenvectors belonging to distinct eigenvalues are linearly independent). Hence the list $e^{\lambda_1 x},\dots,e^{\lambda_n x}$ is linearly independent. ∎

## Exercise 5.A.38

**Statement.**

Suppose $V$ is finite-dimensional, $T \in \mathcal{L}(V)$, and $U$ is a subspace of $V$ invariant under $T$. The *quotient operator* $T/U \in \mathcal{L}(V/U)$ is defined by  

$$
(T/U)(v + U) = Tv + U
$$

for each $v \in V$.

(a) Show that the definition of $T/U$ makes sense (which requires using the condition that $U$ is invariant under $T$) and show that $T/U$ is an operator on $V/U$.

(b) Show that each eigenvalue of $T/U$ is an eigenvalue of $T$.


### Solution

#### Part (a)

**Well‑definedness.**  
Take $u,v \in V$ such that $v+U = u+U$. Then $v-u \in U$. Because $U$ is invariant under $T$, $T(v-u) \in U$. Now compute  

$$
(T/U)(v+U) - (T/U)(u+U) = (Tv+U) - (Tu+U) = (Tv - Tu) + U = T(v-u) + U.
$$

Since $T(v-u) \in U$, we have $T(v-u)+U = U$, the zero element of $V/U$. Therefore  

$$
(T/U)(v+U) = (T/U)(u+U),
$$

so the definition does not depend on the representative of the coset.

**Linearity.**  
For any $v,w \in V$ and $\lambda \in \mathbf{F}$:

$$
\begin{aligned}
(T/U)\bigl((v+U)+(w+U)\bigr) &= (T/U)\bigl((v+w)+U\bigr) = T(v+w)+U = (Tv+Tw)+U \\
&= (Tv+U)+(Tw+U) = (T/U)(v+U) + (T/U)(w+U),
\end{aligned}
$$

$$
\begin{aligned}
(T/U)\bigl(\lambda(v+U)\bigr) &= (T/U)(\lambda v+U) = T(\lambda v)+U = \lambda Tv+U \\
&= \lambda (Tv+U) = \lambda (T/U)(v+U).
\end{aligned}
$$

Hence $T/U$ is a linear operator on $V/U$.

---

#### Part (b)

Let $\lambda$ be an eigenvalue of $T/U$. Then there exists a non‑zero coset $v+U \in V/U$ such that  

$$
(T/U)(v+U) = \lambda (v+U).
$$

By definition, this means  

$$
Tv + U = \lambda v + U \quad\Longrightarrow\quad Tv - \lambda v \in U.
$$

Thus $(T-\lambda I)v \in U$.

Consider the restriction of $T-\lambda I$ to the subspace $U$. We distinguish two cases.

**Case 1:** $T-\lambda I|_U$ is **not** injective.  
Then there exists a non‑zero $u \in U$ with $(T-\lambda I)u = 0$. This $u$ is an eigenvector of $T$ with eigenvalue $\lambda$, so $\lambda$ is an eigenvalue of $T$.

**Case 2:** $T-\lambda I|_U$ is **injective**.  
Because $U$ is finite‑dimensional (since $V$ is finite‑dimensional), injectivity implies surjectivity. Hence $T-\lambda I|_U$ is also surjective onto $U$.

Since $(T-\lambda I)v \in U$, surjectivity gives a vector $w \in U$ such that  

$$
(T-\lambda I)w = (T-\lambda I)v.
$$

Define $v' = v - w$. If $v' = 0$, then $v = w \in U$ (because $w \in U$), which would imply $v+U = U$, contradicting that $v+U$ is non‑zero. Hence $v' \neq 0$. Moreover,

$$
(T-\lambda I)v' = (T-\lambda I)(v-w) = (T-\lambda I)v - (T-\lambda I)w = 0.
$$

Thus $v'$ is a non‑zero vector in $\ker(T-\lambda I)$. Therefore $\lambda$ is an eigenvalue of $T$.

In both cases, $\lambda$ is an eigenvalue of $T$. ∎
