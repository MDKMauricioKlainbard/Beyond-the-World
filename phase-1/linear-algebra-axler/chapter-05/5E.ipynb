## Exercise 5E.10

**Statement.**

Give an example of two commuting operators $S, T$ on a finite-dimensional real vector space such that $S + T$ has an eigenvalue that does not equal an eigenvalue of $S$ plus an eigenvalue of $T$ and $ST$ has an eigenvalue that does not equal an eigenvalue of $S$ times an eigenvalue of $T$.

### Solution

Let $V = \mathbf{R}^2$, which is a finite-dimensional real vector space. We define the linear operators $T, S \in \mathcal{L}(\mathbf{R}^2)$ as follows:
$$
T(x, y) = (-y, x)
$$
$$
S(x, y) = (y, -x)
$$

Geometrically, $T$ represents a counterclockwise rotation of $90^\circ$ in the Cartesian plane, and $S$ represents a clockwise rotation of $90^\circ$.

---

#### 1. Commutativity of $S$ and $T$
We compute the compositions $ST$ and $TS$ for an arbitrary vector $(x, y) \in \mathbf{R}^2$:
$$
(ST)(x, y) = S(T(x, y)) = S(-y, x) = (x, -(-y)) = (x, y)
$$
$$
(TS)(x, y) = T(S(x, y)) = T(y, -x) = (-(-x), y) = (x, y)
$$
Since $ST(x, y) = TS(x, y) = I(x, y)$ for all vectors, the operators commute ($ST = TS = I$).

---

#### 2. Eigenvalues of $S$ and $T$
Let us find the eigenvalues of $T$. Suppose there exists $\lambda \in \mathbf{R}$ and a non-zero vector $(x, y) \in \mathbf{R}^2$ such that $T(x, y) = \lambda (x, y)$. This leads to the system of equations:
1. $-y = \lambda x$
2. $x = \lambda y$

Substituting the second equation into the first gives $-y = \lambda(\lambda y) = \lambda^2 y$, which implies $(\lambda^2 + 1)y = 0$. Since $(x, y) \neq (0, 0)$, if $y = 0$ then $x = 0$, so we must have $y \neq 0$. Thus:
$$
\lambda^2 + 1 = 0 \implies \lambda^2 = -1
$$
Because there is no real number $\lambda \in \mathbf{R}$ that satisfies this equation, $T$ has no eigenvalues. By an identical algebraic calculation, $S$ also has no eigenvalues.

---

#### 3. Eigenvalues of $S + T$ and $ST$
Now we analyze the sum and product operators:

- **The sum operator $S + T$:**
  $$
  (S + T)(x, y) = S(x, y) + T(x, y) = (y, -x) + (-y, x) = (0, 0)
  $$
  Thus, $S + T = 0$ is the zero operator. The zero operator maps every non-zero vector to $0$, meaning that $\lambda = 0$ is an eigenvalue of $S + T$.

- **The product operator $ST$:**
  As shown during the commutativity check, $ST = I$ is the identity operator. The identity operator satisfies $I(x, y) = 1(x, y)$ for all vectors, meaning that $\mu = 1$ is an eigenvalue of $ST$.

---

#### Conclusion
The set of eigenvalues of $S$ is empty ($\emptyset$), and the set of eigenvalues of $T$ is empty ($\emptyset$). 

- The operator $S + T$ has the eigenvalue $0$. Since neither $S$ nor $T$ possesses any real eigenvalues, it is impossible to express $0$ as $\lambda_S + \lambda_T$ for any $\lambda_S, \lambda_T \in \mathbf{R}$.
- The operator $ST$ has the eigenvalue $1$. For the same reason, it is impossible to express $1$ as $\lambda_S \cdot \lambda_T$.

This provides the required counterexample, proving that the property established for complex vector spaces does not extend to real vector spaces. ∎
