## Exercise 1C.1

**Statement.**

For each of the following subsets of $\mathbf{F}^3$, determine whether it is a subspace of $\mathbf{F}^3$:
(a) $\{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 + 2x_2 + 3x_3 = 0\}$
(b) $\{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 + 2x_2 + 3x_3 = 4\}$
(c) $\{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 x_2 x_3 = 0\}$
(d) $\{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 = 5x_3\}$

### Solution

#### (a) Subspace
The subset $U = \{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 + 2x_2 + 3x_3 = 0\}$ is a subspace of $\mathbf{F}^3$.

* **Additive Identity:** Setting $x_1 = x_2 = x_3 = 0$ yields $0 + 2(0) + 3(0) = 0$, so $(0,0,0) \in U$.
* **Closure under Addition:** Let $u = (u_1, u_2, u_3) \in U$ and $v = (v_1, v_2, v_3) \in U$. By definition, $u_1 + 2u_2 + 3u_3 = 0$ and $v_1 + 2v_2 + 3v_3 = 0$. Summing these two equations yields:
  $$(u_1 + v_1) + 2(u_2 + v_2) + 3(u_3 + v_3) = 0$$
  Since the coordinates of $u + v$ satisfy the defining equation, $u + v \in U$.
* **Closure under Scalar Multiplication:** Let $\lambda \in \mathbf{F}$ and $u \in U$. Multiplying the equation $u_1 + 2u_2 + 3u_3 = 0$ by $\lambda$ gives:
  $$(\lambda u_1) + 2(\lambda u_2) + 3(\lambda u_3) = 0$$
  Thus, $\lambda u \in U$.

#### (b) Not a Subspace
The subset $U = \{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 + 2x_2 + 3x_3 = 4\}$ is not a subspace because it fails to contain the additive identity. Substituting the zero vector $(0,0,0)$ into the condition gives $0 + 2(0) + 3(0) = 0 \neq 4$, meaning $(0,0,0) \notin U$.

#### (c) Not a Subspace
The subset $U = \{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 x_2 x_3 = 0\}$ is not a subspace because it is not closed under addition. 

Consider the vectors $u = (1, 1, 0)$ and $v = (0, 1, 1)$. Both vectors belong to $U$ since $1 \cdot 1 \cdot 0 = 0$ and $0 \cdot 1 \cdot 1 = 0$. However, their sum is:
$$u + v = (1, 2, 1)$$
Evaluating the defining condition for the sum yields $1 \cdot 2 \cdot 1 = 2 \neq 0$, which proves that $u + v \notin U$.

#### (d) Subspace
The subset $U = \{(x_1, x_2, x_3) \in \mathbf{F}^3 : x_1 = 5x_3\}$ can be rewritten via the homogeneous linear equation $x_1 - 5x_3 = 0$, making it a subspace of $\mathbf{F}^3$.

* **Additive Identity:** Since $0 = 5(0)$, the zero vector $(0,0,0) \in U$.
* **Closure under Addition:** Let $u, v \in U$, which implies $u_1 = 5u_3$ and $v_1 = 5v_3$. Adding these equations together yields:
  $$(u_1 + v_1) = 5(u_3 + v_3)$$
  Therefore, $u + v \in U$.
* **Closure under Scalar Multiplication:** Let $\lambda \in \mathbf{F}$ and $u \in U$. Multiplying $u_1 = 5u_3$ by $\lambda$ gives:
  $$(\lambda u_1) = 5(\lambda u_3)$$
  Therefore, $\lambda u \in U$. $\blacksquare$

## Exercise 1C.3

**Statement.**

Show that the set of differentiable real-valued functions $f$ on the interval $(-4, 4)$ such that $f'(-1) = 3f(2)$ is a subspace of $\mathbf{R}^{(-4,4)}$.

### Proof

Let $U$ be the set of differentiable real-valued functions $f$ on the interval $(-4, 4)$ satisfying $f'(-1) = 3f(2)$. To prove that $U$ is a subspace of $\mathbf{R}^{(-4,4)}$, we verify the three mandatory criteria:

* **Additive Identity:** Let $0$ denote the zero function on $(-4,4)$, defined by $0(x) = 0$ for all $x \in (-4,4)$. The zero function is differentiable on this interval, and its derivative is also the zero function, meaning $0'(x) = 0$ for all $x$. Evaluating both sides of the defining condition yields:
  $$0'(-1) = 0 \quad \text{and} \quad 3 \cdot 0(2) = 3(0) = 0$$
  Since $0'(-1) = 3 \cdot 0(2)$, the zero function belongs to $U$.

* **Closure under Addition:** Let $f \in U$ and $g \in U$. By definition, both functions are differentiable on $(-4,4)$ and satisfy the conditions $f'(-1) = 3f(2)$ and $g'(-1) = 3g(2)$. The sum of two differentiable functions is differentiable on the same interval. Using the additivity of the derivative operator and functional addition, we evaluate the condition for $f+g$:
  $$(f+g)'(-1) = f'(-1) + g'(-1)$$
  Substituting the structural equations for $f$ and $g$:
  $$(f+g)'(-1) = 3f(2) + 3g(2) = 3\big(f(2) + g(2)\big) = 3(f+g)(2)$$
  Since the coordinates satisfy the governing equation, $f+g \in U$.

* **Closure under Scalar Multiplication:** Let $\lambda \in \mathbf{R}$ and $f \in U$. The scalar multiple of a differentiable function is differentiable on $(-4,4)$. Applying the homogeneity of the derivative operator and functional scalar multiplication:
  $$(\lambda f)'(-1) = \lambda f'(-1)$$
  Substituting the hypothesis $f'(-1) = 3f(2)$:
  $$(\lambda f)'(-1) = \lambda \big(3f(2)\big) = 3\big(\lambda f(2)\big) = 3(\lambda f)(2)$$
  Therefore, $\lambda f \in U$.

Since $U$ contains the additive identity and is closed under both addition and scalar multiplication, it is a subspace of $\mathbf{R}^{(-4,4)}$. $\blacksquare$

## Exercise 1C.4

**Statement.**

Suppose $b \in \mathbf{R}$. Show that the set of continuous real-valued functions $f$ on the interval $[0, 1]$ such that $\int_0^1 f = b$ is a subspace of $\mathbf{R}^{[0,1]}$ if and only if $b = 0$.

### Proof

Let $U = \{f \in \mathbf{R}^{[0,1]} : f \text{ is continuous and } \int_0^1 f = b\}$. We need to prove that $U$ is a subspace of $\mathbf{R}^{[0,1]}$ if and only if $b = 0$.

#### Forward Direction ($\Rightarrow$)
Assume $U$ is a subspace of $\mathbf{R}^{[0,1]}$. 
* Every subspace must contain the additive identity (the zero function $0$).
* Therefore, $0 \in U$, which implies that the zero function must satisfy the defining integral condition of the set:
  $$\int_0^1 0 \, dx = b$$
* Since the definite integral of the zero function over any interval is $0$, it follows immediately that $0 = b$.

#### Backward Direction ($\Leftarrow$)
Assume $b = 0$. The subset becomes $U = \{f \in \mathbf{R}^{[0,1]} : f \text{ is continuous and } \int_0^1 f = 0\}$. We verify the three subspace conditions:

* **Additive Identity:** The zero function $0$ is continuous on $[0,1]$ and satisfies $\int_0^1 0 \, dx = 0$, so $0 \in U$.
* **Closure under Addition:** Let $f \in U$ and $g \in U$. This means both functions are continuous and $\int_0^1 f = 0$ and $\int_0^1 g = 0$. The sum $f+g$ is continuous, and by the linearity of integration:
  $$\int_0^1 (f+g) \, dx = \int_0^1 f \, dx + \int_0^1 g \, dx = 0 + 0 = 0$$
  Thus, $f+g \in U$.
* **Closure under Scalar Multiplication:** Let $\lambda \in \mathbf{R}$ and $f \in U$. The scalar multiple $\lambda f$ is continuous, and by the homogeneity of integration:
  $$\int_0^1 (\lambda f) \, dx = \lambda \int_0^1 f \, dx = \lambda(0) = 0$$
  Thus, $\lambda f \in U$.

Since all conditions hold when $b = 0$, $U$ is a subspace. $\blacksquare$

## Exercise 1C.10

**Statement.**

Suppose $V_1$ and $V_2$ are subspaces of $V$. Prove that the intersection $V_1 \cap V_2$ is a subspace of $V$.

### Proof

Let $V_1$ and $V_2$ be subspaces of $V$. To prove that the intersection $V_1 \cap V_2$ is a subspace of $V$, we verify the three mandatory criteria:

* **Additive Identity:** Since $V_1$ is a subspace of $V$, it must contain the additive identity, hence $0 \in V_1$. Similarly, because $V_2$ is a subspace of $V$, it follows that $0 \in V_2$. Since the zero vector belongs to both individual subspaces, it belongs to their intersection:
  $$0 \in V_1 \cap V_2$$

* **Closure under Addition:** Let $u, v \in V_1 \cap V_2$. By definition of set intersection, this implies that $u, v \in V_1$ and $u, v \in V_2$. 
  * Since $V_1$ is a subspace, it is closed under addition, meaning $u + v \in V_1$.
  * Since $V_2$ is a subspace, it is also closed under addition, meaning $u + v \in V_2$.
  As the sum vector $u + v$ satisfies the membership condition for both sets simultaneously, we conclude that $u + v \in V_1 \cap V_2$.

* **Closure under Scalar Multiplication:** Let $\lambda \in \mathbf{F}$ and $u \in V_1 \cap V_2$. The definition of intersection implies that $u \in V_1$ and $u \in V_2$.
  * Because $V_1$ is a subspace, it is closed under scalar multiplication, which yields $\lambda u \in V_1$.
  * Because $V_2$ is a subspace, it is likewise closed under scalar multiplication, yielding $\lambda u \in V_2$.
  Therefore, the scalar multiple $\lambda u$ is an element of both subspaces, establishing that $\lambda u \in V_1 \cap V_2$.

Since $V_1 \cap V_2$ contains the additive identity and is closed under both addition and scalar multiplication, it is a subspace of $V$. $\blacksquare$

## Exercise 1C.15

**Statement.**

Suppose $U$ is a subspace of $V$. What is $U + U$?

### Solution

The sum of the subspace with itself yields $U + U = U$. We prove this equality by demonstrating mutual subset containment:

* **Containment $U \subseteq U + U$:** Let $x \in U$. Since $U$ is a subspace of $V$, it must contain the additive identity, meaning $0 \in U$. We can express $x$ as a sum of two elements from $U$:
  $$x = x + 0$$
  By the definition of subspace addition, because $x \in U$ and $0 \in U$, their sum belongs to $U + U$. Therefore, $U \subseteq U + U$.

* **Containment $U + U \subseteq U$:** Let $w \in U + U$. By the definition of the sum of subspaces, there exist vectors $x, y \in U$ such that $w = x + y$. Since $U$ is a subspace, it is strictly closed under addition. This ensures that the sum of any two elements in $U$ remains within $U$, implying that $x + y \in U$. Consequently, $w \in U$, establishing that $U + U \subseteq U$.

Since both containments hold simultaneously, we conclude that $U + U = U$. $\blacksquare$
