## Exercise 3D.1

**Statement.**

Suppose $T \in \mathcal{L}(V, W)$ is invertible. Show that $T^{-1}$ is invertible and $(T^{-1})^{-1} = T$.

### Proof

By definition, since $T \in \mathcal{L}(V, W)$ is invertible, there exists a unique linear map $T^{-1} \in \mathcal{L}(W, V)$ such that:
$$T T^{-1} = I_W \quad \text{and} \quad T^{-1} T = I_V$$

where $I_W$ and $I_V$ denote the identity maps on $W$ and $V$, respectively.

To prove that $T^{-1}$ is invertible, we must show there exists a linear map $R \in \mathcal{L}(V, W)$ satisfying the criteria for the inverse of $T^{-1}$, namely:
$$T^{-1} R = I_V \quad \text{and} \quad R T^{-1} = I_W$$

Choosing $R = T$ and substituting it into the expressions yields:
$$T^{-1} T = I_V \quad \text{and} \quad T T^{-1} = I_W$$

Both equations are identical to the definition of the invertibility of $T$ and are therefore true. This establishes that $T^{-1}$ is invertible and that its unique inverse is $T$, meaning:
$$(T^{-1})^{-1} = T \quad \blacksquare$$

## Exercise 3D.3

**Statement.**

Suppose $V$ is finite-dimensional and $T \in \mathcal{L}(V)$. Prove that the following are equivalent:
1. $T$ is invertible.
2. $T v_1, \dots, T v_n$ is a basis of $V$ for every basis $v_1, \dots, v_n$ of $V$.
3. $T v_1, \dots, T v_n$ is a basis of $V$ for some basis $v_1, \dots, v_n$ of $V$.

### Proof

Let $\dim V = n$. We will prove the equivalence by demonstrating the implications $(1) \implies (2)$, $(2) \implies (3)$, and $(3) \implies (1)$.

#### Part 1: $(1) \implies (2)$
Assume $T$ is invertible, which implies $T$ is injective. Let $v_1, \dots, v_n$ be an arbitrary basis of $V$. To show that $T v_1, \dots, T v_n$ is a basis of $V$, we first check its linear independence. Let $c_1, \dots, c_n \in \mathbf{F}$ be scalars such that:
$$c_1 T v_1 + \dots + c_n T v_n = 0$$

By the linearity of $T$, this can be written as:
$$T(c_1 v_1 + \dots + c_n v_n) = 0$$

Since $T$ is injective, its null space is trivial, meaning the argument must be the zero vector:
$$c_1 v_1 + \dots + c_n v_n = 0$$

Because $v_1, \dots, v_n$ is a basis, it is linearly independent, which forces $c_1 = \dots = c_n = 0$. Thus, the list $T v_1, \dots, T v_n$ is linearly independent. Since its length is $n = \dim V$, it automatically constitutes a basis of $V$.

#### Part 2: $(2) \implies (3)$
This implication is immediate. If $T v_1, \dots, T v_n$ is a basis for *every* basis of $V$, then it is necessarily a basis for *some* specific basis of $V$.

#### Part 3: $(3) \implies (1)$
Assume there exists some basis $v_1, \dots, v_n$ of $V$ such that $T v_1, \dots, T v_n$ is a basis of $V$. We must show that $T$ is both surjective and injective.

* **Surjectivity:** Since $T v_1, \dots, T v_n$ is a basis of $V$, it spans $V$. Because every element in this list belongs to $\text{range } T$, it follows that $\text{range } T = V$. Thus, $T$ is surjective.
* **Injectivity:** Let $u \in \text{null } T$. Since $v_1, \dots, v_n$ is a basis of $V$, we can express $u$ as a linear combination of these vectors for some scalars $a_1, \dots, a_n \in \mathbf{F}$:
  $$u = a_1 v_1 + \dots + a_n v_n$$
  Applying $T$ to both sides yields:
  $$T u = T(a_1 v_1 + \dots + a_n v_n) = a_1 T v_1 + \dots + a_n T v_n$$
  Since $u \in \text{null } T$, we have $T u = 0$, so:
  $$a_1 T v_1 + \dots + a_n T v_n = 0$$
  By hypothesis, $T v_1, \dots, T v_n$ is a basis, making it linearly independent. Therefore, $a_1 = \dots = a_n = 0$. Substituting these coefficients back into the expression for $u$ gives $u = 0$. Hence, $\text{null } T = \{0\}$, proving $T$ is injective.

Since $T$ is both injective and surjective on a finite-dimensional space, $T$ is invertible $\quad \blacksquare$

## Exercise 3D.4

**Statement.**

Suppose $V$ is finite-dimensional and $\dim V > 1$. Prove that the set of noninvertible linear maps from $V$ to itself is not a subspace of $\mathcal{L}(V)$.

### Proof

Let $\dim V = n$ with $n > 1$, and let $v_1, \dots, v_n$ be a basis of $V$. To prove that the set of noninvertible linear maps is not a subspace of $\mathcal{L}(V)$, it suffices to show that the set is not closed under addition.

We define a linear map $T \in \mathcal{L}(V)$ by specifying its action on the basis elements:
$$T(v_k) = v_k \quad \text{for } k = 1, \dots, n-1$$
$$T(v_n) = 0$$

The list of images $T v_1, \dots, T v_n$ contains the zero vector, making the list linearly dependent. Because $T$ maps a basis to a linearly dependent list, $T$ is not invertible.

Next, we define another linear map $S \in \mathcal{L}(V)$ by:
$$S(v_k) = 0 \quad \text{for } k = 1, \dots, n-1$$
$$S(v_n) = v_n$$

Since $n > 1$, the list of images $S v_1, \dots, S v_n$ contains at least one zero vector (specifically $S v_1 = 0$), meaning this list is also linearly dependent. Thus, $S$ is not invertible.

Now, consider the sum of these two maps, $(T+S) \in \mathcal{L}(V)$. Evaluating the action of this combined operator on each basis vector yields:
$$(T+S)(v_k) = T(v_k) + S(v_k) = v_k + 0 = v_k \quad \text{for } k = 1, \dots, n-1$$
$$(T+S)(v_n) = T(v_n) + S(v_n) = 0 + v_n = v_n$$

This shows that $(T+S)(v_k) = v_k$ for all $k = 1, \dots, n$. The list of images under $T+S$ is exactly the original basis $v_1, \dots, v_n$, which is linearly independent and spans $V$. This implies that $T+S$ is the identity map $I$, which is invertible.

Since both $T$ and $S$ belong to the set of noninvertible linear maps, but their sum $T+S$ is invertible, the set is not closed under addition. Consequently, the set of noninvertible linear maps from $V$ to itself is not a subspace of $\mathcal{L}(V) \quad \blacksquare$

## Exercise 3D.9

**Statement.**

Suppose $V$ is finite-dimensional and $T: V \to W$ is a surjective linear map of $V$ onto $W$. Prove that there is a subspace $U$ of $V$ such that $T|_U$ is an isomorphism of $U$ onto $W$.

### Proof

Let $u_1, \dots, u_m$ be a basis of $\text{null } T$. Since $V$ is finite-dimensional, we can extend this list to a basis of $V$:
$$u_1, \dots, u_m, v_1, \dots, v_n$$

We define the subspace $U \subseteq V$ as the span of the extended vectors:
$$U = \text{span}(v_1, \dots, v_n)$$

To prove that the restriction $T|_U: U \to W$ is an isomorphism, we demonstrate that it is both injective and surjective.

**Injectivity:**
Suppose $u \in U$ satisfies $(T|_U)(u) = 0$. By definition, this means $T(u) = 0$, so $u \in \text{null } T$. Since $u \in U$, it can be written as a linear combination of the vectors $v_1, \dots, v_n$. Simultaneously, since $u \in \text{null } T$, it can be expressed as a linear combination of the kernel basis $u_1, \dots, u_m$:
$$u = \sum_{j=1}^n c_j v_j = \sum_{i=1}^m d_i u_i \implies \sum_{i=1}^m d_i u_i - \sum_{j=1}^n c_j v_j = 0$$

Because the combined list $u_1, \dots, u_m, v_1, \dots, v_n$ is a basis of $V$, it is linearly independent. Consequently, all coefficients must equal zero, yielding $c_j = 0$ for all $j$, which implies $u = 0$. Thus, $\text{null } T|_U = \{0\}$, establishing that $T|_U$ is injective.

**Surjectivity:**
Let $w \in W$. Since $T$ is surjective, there exists a vector $v \in V$ such that $T(v) = w$. Expressing $v$ in terms of the full basis of $V$ gives:
$$v = \sum_{i=1}^m a_i u_i + \sum_{j=1}^n b_j v_j$$

Applying $T$ to both sides, and noting that $T(u_i) = 0$ for all $i \in \{1, \dots, m\}$, we obtain:
$$T(v) = \sum_{i=1}^m a_i T(u_i) + T\left(\sum_{j=1}^n b_j v_j\right) = 0 + T\left(\sum_{j=1}^n b_j v_j\right)$$

Let $u_0 = \sum_{j=1}^n b_j v_j$. By construction, $u_0 \in U$. This means:
$$(T|_U)(u_0) = T(u_0) = T(v) = w$$

Since for every $w \in W$ there exists an element $u_0 \in U$ such that $(T|_U)(u_0) = w$, the restriction is surjective.

Because $T|_U: U \to W$ is both injective and surjective, it is an isomorphism of $U$ onto $W \quad \blacksquare$

## Exercise 3.D.10

**Statement.**

Suppose $V$ and $W$ are finite‑dimensional and $U$ is a subspace of $V$. Let  

$$
\mathcal{E} = \{ T \in \mathcal{L}(V, W) : U \subseteq \text{null } T \}.
$$

(a) Show that $\mathcal{E}$ is a subspace of $\mathcal{L}(V, W)$.  

(b) Find a formula for $\dim \mathcal{E}$ in terms of $\dim V$, $\dim W$, and $\dim U$.

*Hint: Define $\Phi : \mathcal{L}(V, W) \to \mathcal{L}(U, W)$ by $\Phi(T) = T|_U$. What is null $\Phi$? What is range $\Phi$?*


### Solution

#### Part (a)

We verify the three subspace criteria.

- **Zero map:** The zero map $0 \in \mathcal{L}(V,W)$ clearly satisfies $U \subseteq \operatorname{null} 0$, so $0 \in \mathcal{E}$.

- **Closure under addition:** Take $T, S \in \mathcal{E}$. For any $u \in U$,
  $$
  (T+S)(u) = T(u) + S(u) = 0 + 0 = 0,
  $$
  hence $U \subseteq \operatorname{null}(T+S)$. Thus $T+S \in \mathcal{E}$.

- **Closure under scalar multiplication:** Let $T \in \mathcal{E}$ and $\lambda \in \mathbf{F}$. For $u \in U$,
  $$
  (\lambda T)(u) = \lambda T(u) = \lambda \cdot 0 = 0,
  $$
  so $U \subseteq \operatorname{null}(\lambda T)$. Therefore $\lambda T \in \mathcal{E}$.

Thus $\mathcal{E}$ is a subspace of $\mathcal{L}(V,W)$.

---

#### Part (b)

Define the restriction map  

$$
\Phi : \mathcal{L}(V,W) \longrightarrow \mathcal{L}(U,W), \qquad \Phi(T) = T|_U.
$$

It is easy to see that $\Phi$ is linear (restriction preserves addition and scalar multiplication).

- **Kernel of $\Phi$:**  
  $T \in \operatorname{null} \Phi$ iff $T|_U = 0$, i.e. $T(u)=0$ for every $u \in U$. This is exactly $U \subseteq \operatorname{null} T$. Hence  
  $$
  \operatorname{null} \Phi = \mathcal{E}.
  $$

- **Range of $\Phi$:**  
  We claim $\Phi$ is surjective. Let $\psi \in \mathcal{L}(U,W)$. Choose a basis $\{u_1,\dots,u_m\}$ of $U$ (where $m = \dim U$) and extend it to a basis $\{u_1,\dots,u_m, v_{m+1},\dots,v_n\}$ of $V$ (where $n = \dim V$). Define $T \in \mathcal{L}(V,W)$ by setting $T(u_i) = \psi(u_i)$ for $i=1,\dots,m$ and $T(v_j) = 0$ for $j=m+1,\dots,n$, then extending linearly. Then $\Phi(T) = T|_U = \psi$. Therefore $\operatorname{range} \Phi = \mathcal{L}(U,W)$.

Because $V$ and $W$ are finite‑dimensional, $\mathcal{L}(V,W)$ and $\mathcal{L}(U,W)$ are also finite‑dimensional. Apply the rank‑nullity theorem to $\Phi$:

$$
\dim \mathcal{L}(V,W) = \dim \operatorname{null} \Phi + \dim \operatorname{range} \Phi.
$$

We know:
- $\dim \mathcal{L}(V,W) = (\dim V)(\dim W)$,
- $\dim \operatorname{null} \Phi = \dim \mathcal{E}$,
- $\dim \operatorname{range} \Phi = \dim \mathcal{L}(U,W) = (\dim U)(\dim W)$.

Hence  

$$
(\dim V)(\dim W) = \dim \mathcal{E} + (\dim U)(\dim W).
$$

Solving for $\dim \mathcal{E}$:

$$
\dim \mathcal{E} = (\dim V)(\dim W) - (\dim U)(\dim W) = (\dim V - \dim U)\,\dim W.
$$

∎

## Exercise 3D.21

**Statement.**

Suppose $n$ is a positive integer and $A_{j,k} \in \mathbf{F}$ for all $j, k = 1, \dots, n$. Prove that the following are equivalent:
(a) The trivial solution $x_1 = \dots = x_n = 0$ is the only solution to the homogeneous system of equations
$$\sum_{k=1}^n A_{1,k} x_k = 0$$
$$\vdots$$
$$\sum_{k=1}^n A_{n,k} x_k = 0.$$

(b) For every $c_1, \dots, c_n \in \mathbf{F}$, there exists a solution to the system of equations
$$\sum_{k=1}^n A_{1,k} x_k = c_1$$
$$\vdots$$
$$\sum_{k=1}^n A_{n,k} x_k = c_n.$$

### Proof

We define a linear map $T: \mathbf{F}^n \to \mathbf{F}^n$ by:
$$T(x_1, \dots, x_n) = \left( \sum_{k=1}^n A_{1,k} x_k, \dots, \sum_{k=1}^n A_{n,k} x_k \right)$$

The linearity of $T$ follows directly from the algebraic properties of vector addition and scalar multiplication in $\mathbf{F}$. We can now rephrase statements (a) and (b) in terms of the properties of this linear operator:

- Statement (a) asserts that $T(x_1, \dots, x_n) = (0, \dots, 0)$ implies $(x_1, \dots, x_n) = (0, \dots, 0)$. This is exactly the definition of injectivity, meaning that $\text{null } T = \{(0, \dots, 0)\}$, which implies $\dim \text{null } T = 0$.

- Statement (b) asserts that for every vector $(c_1, \dots, c_n) \in \mathbf{F}^n$, there exists an element $(x_1, \dots, x_n) \in \mathbf{F}^n$ such that $T(x_1, \dots, x_n) = (c_1, \dots, c_n)$. This is exactly the definition of surjectivity, meaning that $\text{range } T = \mathbf{F}^n$.

By the Fundamental Theorem of Linear Maps, we have:
$$\dim \text{null } T + \dim \text{range } T = \dim \mathbf{F}^n = n$$

If (a) holds, then $\dim \text{null } T = 0$. Substituting this into the equation yields $\dim \text{range } T = n$. Since $\text{range } T$ is a subspace of $\mathbf{F}^n$ and has the same dimension $n$, it must be that $\text{range } T = \mathbf{F}^n$, which proves (b).

Conversely, if (b) holds, then $\text{range } T = \mathbf{F}^n$, so $\dim \text{range } T = n$. Substituting this into the equation yields $\dim \text{null } T = 0$, which implies $\text{null } T = \{(0, \dots, 0)\}$, proving (a).

Thus, statement (a) and statement (b) are equivalent $\quad \blacksquare$
